### 两数之和
给定一个整数数组 nums 和一个整数目标值 target，请你在该数组中找出 和为目标值 target  的那 两个 整数，并返回它们的数组下标。

你可以假设每种输入只会对应一个答案，并且你不能使用两次相同的元素。

你可以按任意顺序返回答案。

思路：用一个哈希表记录每个数字对应的下标，然后在遍历数组时判断 target - nums[i] 是否在哈希表中，如果在则找到了答案。

实现：
```cpp
#include <vector>
#include <unordered_map>
#include <iostream>
#include <algorithm>
using namespace std;

vector<int> twoSum(vector<int>& nums, int target) {
    unordered_map<int, int> mp;
    for(int i = 0; i < nums.size(); i++){
        if(mp.count(target - nums[i])) return {mp[target - nums[i]], i};
        mp[nums[i]] = i;
    }
    return {};
}

int main(){
    vector<int> nums = {2, 7, 11, 15};
    int target = 9;
    vector<int> ans = twoSum(nums, target);
    cout << ans[0] << " " << ans[1] << "\n";
    return 0;
}
```

### 字母异位次分组
给定一个字符串数组，将字母异位词组合在一起。字母异位词指字母相同，但排列不同的字符串。

思路：先对每个字符串进行排序，然后将排序后的字符串作为哈希表的键，将原字符串加入对应的值列表中。

注意：排序后的字符串作为键时，可能会有重复的键，所以需要使用一个数组来存储对应的原字符串。

实现：
```cpp
#include <vector>
#include <string>
#include <unordered_map>
#include <iostream>
#include <algorithm>

using namespace std;

vector<vector<string>> groupAnagrams(vector<string>& strs) {
    unordered_map<string, vector<string>> mp;
    for(int i = 0; i < strs.size(); i++){
        string s = strs[i];
        sort(s.begin(), s.end());
        mp[s].push_back(strs[i]);
    }
    vector<vector<string>> ans;
    for(const auto& it : mp){
        ans.push_back(it.second);
    }
    return ans;
}

int main(){
    vector<string> strs = {"eat", "tea", "tan", "ate", "nat", "bat"};
    vector<vector<string>> ans = groupAnagrams(strs);
    for(const auto& group : ans){
        for(const auto& str : group){
            cout << str << " ";
        }
        cout << "\n";
    }
    return 0;
}
```

### 最长连续序列
给定一个未排序的整数数组 nums ，找出数字连续的最长序列（不要求序列元素在原数组中连续）的长度。

请你设计并实现时间复杂度为 O(n) 的算法解决此问题。

思路：将数组中的元素存入一个哈希表中，然后遍历哈希表中的每个元素，如果该元素的前一个元素不在哈希表中，则说明该元素是一个连续序列的起点，然后不断向后查找连续的元素，直到找不到为止，记录下连续序列的长度，更新答案。

实现：
```cpp
#include <vector>
#include <unordered_set>
#include <algorithm>
#include <iostream>
using namespace std;

int longestConsecutive(vector<int>& nums) {
    unordered_set<int> st(nums.begin(), nums.end());
    int ans = 0;
    for(int x : st){
        if(st.count(x - 1)) continue; //核心逻辑，如果x-1在哈希表中，说明x不是连续序列的起点，跳过
        int y = x + 1;
        while(st.count(y)) y++;       //不断向后查找连续的元素，直到找不到为止
        ans = max(ans, y - x);
    }
    return ans;
}

int main(){
    vector<int> nums = {100, 4, 200, 1, 3, 2};
    int ans = longestConsecutive(nums);
    cout << ans << "\n";
    return 0;
}
```

### 移动零
给定一个数组 nums，编写一个函数将所有 0 移动到数组的末尾，同时保持非零元素的相对顺序。

思路：维护一个指针 j，表示当前非零元素应该放置的位置。遍历数组，如果当前元素不为0，就将其放到 j 位置，并将 j 后移。最后将 j 位置及之后的元素都置为0。

实现：
```cpp
#include <vector>
#include <iostream>
#include <algorithm>
using namespace std;

void moveZeroes(vector<int>& nums) {
    int j = 0;
    for(int x : nums){
        if(x) nums[j++] = x;
    }
    fill(nums.begin() + j, nums.end(), 0);
}

int main(){
    vector<int> nums = {0, 1, 0, 3, 12};
    moveZeroes(nums);
    for(int x : nums) cout << x << " ";
    cout << "\n";
    return 0;
}
```

### 盛水最多的容器
给定 n 个非负整数 a1，a2，...，an，每个数代表坐标中的一个点 (i, ai)。在坐标内画 n 条垂直线，垂直线 i 的两个端点分别为 (i, ai) 和 (i, 0)。找出其中的两条线，使得它们与 x 轴共同构成的容器可以容纳最多的水。

思路：对于任意两个索引，盛水的多少取决于较短的那条线。使用双指针，初始时指向数组的两端。计算当前容器的容量，并更新最大容量。然后移动较短的指针，因为移动较长的指针不可能增加容量。

实现：
```cpp
#include <vector>
#include <iostream>
using namespace std;

int maxArea(vector<int>& height) {
    int l = 0, r = height.size() - 1, ans = 0;
    //盛水的面积取决于两个边界较短的那个，所以每次移动较短的边界
    while(l < r){
        int area = 0;
        area = (r - l) * min(height[l], height[r]);
        if(height[l] < height[r]) l++;
        else r--;
        ans = max(ans, area);
    }
    return ans;
};

int main(){
    vector<int> height = {1,8,6,2,5,4,8,3,7};
    cout << maxArea(height) << "\n";
    return 0;
}
```

### 三数之和
给你一个包含 n 个整数的数组 nums，判断 nums 中是否存在三个元素 a，b，c ，使得 a + b + c = 0？请你找出所有满足条件且不重复的三元组。

思路：先将数组排序，然后使用双指针来寻找满足条件的三元组。对于每个元素 nums[i]，使用两个指针分别指向 i+1 和数组末尾，计算三数之和。如果和为0，则找到一个三元组；如果和小于0，则左指针右移；如果和大于0，则右指针左移。

注意：去重操作。因为已经排序，对于i，如果 nums[i] 与 nums[i-1] 相同，则跳过；对于左指针和右指针，如果当前元素与前一个元素相同，也要跳过。

实现：
```cpp
#include <vector>
#include <algorithm>
#include <iostream>
using namespace std;

vector<vector<int>> threeSum(vector<int>& nums) {
    int n = nums.size();
    sort(nums.begin(), nums.end());
    vector<vector<int>> ans;
    for(int i = 0; i < n - 2; i++){
        int x = nums[i];
        if(i > 0 && x == nums[i - 1]) continue; 
        //由于排序了，所以相同的数一定在一起，跳过重复的数  
        int j = i + 1, k = n - 1;
        while(j < k){
            int sum = x + nums[j] + nums[k];
            if(sum == 0){
                ans.push_back({x, nums[j], nums[k]});
                j++;
                k--;
                while(j < k && nums[j] == nums[j - 1]) j++;//去重
                while(k > j && nums[k] == nums[k + 1]) k--;//去重
            }else if(sum < 0) j++;
            else k--;
        }
    }
    return ans;
}

int main(){
    vector<int> nums = {-1, 0, 1, 2, -1, -4};
    vector<vector<int>> ans = threeSum(nums);
    for(auto& v : ans){
        for(auto& x : v) cout << x << " ";
        cout << endl;
    }
    return 0;
}
```

### 接雨水
给定 n 个非负整数表示每个宽度为 1 的柱子的高度图，计算按此排列的柱子，下雨之后能接多少雨水。

思路：同时维护两个数组pre_max和suf_max，pre_max[i]表示i位置左边柱子的最大高度，suf_max[i]表示i位置右边柱子的最大高度。对于每个位置i，能接的雨水量为 min(pre_max[i], suf_max[i]) - height[i]，如果这个值为负数，则说明该位置不能接雨水。

实现：
```cpp
#include <vector>
#include <iostream>

using namespace std;

int trap(vector<int>& height) {
    int n = height.size();
    int ans = 0;
    //同时维护前缀最大值和后缀最大值
    vector<int> pre_max(n);
    pre_max[0] = height[0];
    for(int i = 1; i < n; i++){
        pre_max[i] = max(pre_max[i - 1], height[i]);
    }
    vector<int> suf_max(n);
    suf_max[n - 1] = height[n - 1];
    for(int i = n - 2; i >= 0; i--){
        suf_max[i] = max(suf_max[i + 1], height[i]);
    }
    //计算每个位置能存多少水
    for(int i = 0; i < n; i++){
        ans += min(pre_max[i], suf_max[i]) - height[i];
    }
    return ans;
}

int main(){
    vector<int> height = {0,1,0,2,1,0,1,3,2,1,2,1};
    int ans = trap(height);
    cout << ans << "\n";
    return 0;
}
```

## 滑动窗口

### 无重复字符串的最长字串
给定一个字符串 s ，请你找出其中不含有重复字符的 最长子串 的长度。

思路：使用一个哈希表记录每个字符上次出现的位置，维护一个左指针 l，初始时指向字符串的开头。用右指针遍历字符串，如果当前字符在哈希表中出现过，并且上次出现的位置不小于 l，则说明当前字符与之前的字符重复了，需要将 l 移动到上次出现位置的下一位。每次更新答案为当前r 与 l 之间的距离。

实现：
```cpp
#include <vector>
#include <unordered_map>
#include <iostream>
using namespace std;

int lengthOfLongestSubstring(string s) {
    int ans = 0;
    int n = s.length();
    int l = 0;
    unordered_map<char, int> mp;
    // 右指针不断向右移动，直到遇到重复的字符
    for(int r = 0; r < n; r++){
        mp[s[r]]++;
        if(mp[s[r]] > 1){// 如果遇到重复的字符，左指针不断向右移动，直到不再有重复的字符
            while(mp[s[r]] > 1){
                mp[s[l]]--;
                l++;
            }
        }
        ans = max(ans, r - l + 1);// 更新答案
    }
    return ans;
}

int main(){
    string s = "abcabcbb";
    cout << lengthOfLongestSubstring(s) << endl;
    return 0;
}
```

### 找到字符串中所有字母异位词
给定两个字符串 s 和 p，找到 s 中所有 p 的字母异位词的起始索引。字符串只包含小写英文字母，并且字符串 s 和 p 的长度都不超过 20,100。

思路：先用一个数组记录p中每个字符出现的次数，然后使用一个滑动窗口在s中遍历，每次移动右指针r时更新窗口中字符的出现次数，左指针为l，如果当前窗口中的字符出现次数与p中字符出现次数相同，则说明找到了一个字母异位词，记录下起始索引l。

实现：
```cpp
#include <vector>
#include <iostream>
using namespace std;

vector<int> findAnagrams(string s,string p){
    int n = p.length(), m = s.length();
    vector<int> ans;
    vector<char> array1(26, 0);
    for(char c : p) array1[c- 'a']++;
    vector<char> array2(26, 0);
    for(int r = 0; r < m; r++){
        array2[s[r] - 'a']++;
        int l = r - n + 1;
        if(l < 0) continue;
        if(array1 == array2) ans.push_back(l);
        array2[s[l] - 'a']--;
    }
    return ans;
}

int main(){
    string s = "cbaebabacd", p = "abc";
    vector<int> ans = findAnagrams(s, p);
    for(int i : ans) cout << i << " ";
    cout << endl;
    return 0;
}
```

### 和为K的子数组
给你一个整数数组 nums 和一个整数 k ，请你统计并返回 该数组中和为 k 的子数组的个数 。

子数组是数组中元素的连续非空序列。

思路：将原数组转换为前缀和数组 s，其中 s[i] 表示 nums[0] 到 nums[i-1] 的和，然后对于每个位置 i，计算 s[i] - k 是否在哈希表中出现过，如果出现过则说明存在一个子数组的和为 k。

注意：这里要求的是***连续***的子数组，而不是子序列，所以必须保证子数组中的元素在原数组中是连续的，而不能跳过数组中的元素，所以可以用前缀和+哈希表的方法来统计连续子数组的和。

实现：
```cpp
#include <vector>
#include <unordered_map>
#include <iostream>
using namespace std;

int subarraySum(vector<int>& nums, int k) {
    int n = nums.size();
    vector<int> s(n + 1);
    for(int i = 0; i < n; i++) s[i + 1] = s[i] + nums[i];//前缀和
    unordered_map<int, int> cnt;    //key为和，value为出现次数
    int ans = 0;
    for(int sj : s){
        if(cnt.count(sj - k)) ans += cnt[sj - k];
        //如果存在前缀和为sj-k的情况，则说明存在一个子数组的和为k
        cnt[sj]++;
    }
    return ans;
}

int main(){
    vector<int> nums = {1, 2, 3};
    int k = 3;
    int ans = subarraySum(nums, k);
    cout << ans << "\n";
    return 0;
}
```

### 滑动窗口最大值
给你一个整数数组 nums，有一个大小为 k 的滑动窗口从数组的最左侧移动到数组的最右侧。你只能看到在滑动窗口内的 k 个数字。滑动窗口每次只向右移动一位。请你返回滑动窗口中的最大值。

思路：双端队列。该队列保存nums数组元素的下标，规则为总是保持队列头部的元素是当前窗口的最大值下标，同时队列中元素对应的值从队头到队尾是单调递减的，每次新元素入队时，将比新元素小的元素从队尾弹出，保证队列单调递减，并且每次滑动窗口移动时，如果队头元素已经不在窗口范围内，则将队头弹出。当窗口左端点开始合法时，将队头元素对应的值加入结果数组，即为当前窗口的最大值。

实现：
```cpp
#include <vector>
#include <deque>
#include <iostream>
using namespace std;

vector<int> maxSlidingWindow(vector<int>& nums, int k) {
    int n = nums.size();
    vector<int> ans;
    deque<int> q;
    //队列里存储元素的下标，保证队列头部是当前窗口的最大值
    for(int i = 0; i < n; i++){
        while(!q.empty() && nums[i] >= nums[q.back()]) q.pop_back();
        //比较大小，保持单调递减，同时将较小的元素弹出
        q.push_back(i);
        int l = i - k + 1;
        if(q.front() < l) q.pop_front();
        //如果队列头部的元素已经不在当前窗口了，就弹出
        if(l >= 0) ans.push_back(nums[q.front()]);
    }
    return ans;
}

int main(){
    vector<int> nums = {1,3,-1,-3,5,3,6,7};
    int k = 3;
    vector<int> ans = maxSlidingWindow(nums, k);
    for(const auto& x : ans) cout << x << " ";
    cout << "\n";
    return 0;
}
```

### 最小覆盖子串
给定两个字符串 s 和 t，长度分别是 m 和 n，返回 s 中的 最短窗口 子串，使得该子串包含 t 中的每一个字符（包括重复字符）。如果没有这样的子串，返回空字符串 ""。

思路：核心思路是滑动窗口，通过维护一个窗口 [left, right]，不断向右扩展 right 直到窗口包含 t 中所有字符，然后尝试收缩 left 以找到最短的合法窗口，同时更新最短子串的起始位置和长度。在这个过程中，需要额外的几个变量来判断窗口是否包含 t 中所有字符，包括一个哈希表need记录 t 中每个字符的出现次数，一个哈希表winodw记录当前窗口中每个字符的出现次数，以及一个计数器valid记录当前窗口中已经满足 t 中字符要求的字符个数。right每次向右移动，将 s[right] 加入窗口，同时更新 window 和 valid，当 valid 等于 need 中字符种类数时，说明当前窗口已经包含 t 中所有字符，此时尝试收缩 left，将s[left]移出窗口，同时更新window和valid，直到valid 小于 need中的字符种类。

实现：
```cpp
#include <vector>
#include <unordered_map>
#include <iostream>
#include <climits>
using namespace std;

string minWindow(string s, string t){
    unordered_map<char, int> need, window;
    for(char c : t) need[c]++;
    int valid = 0;
    int left = 0, right = 0;
    int start = 0, len = INT_MAX;

    while(right < s.size()){
        char c = s[right];
        right++;
        if(need.count(c)){
            window[c]++;
            if(window[c] == need[c]) valid++;
        }
        while(valid == need.size()){
            if(start - left < len){
                start = left;
                len = right - left;
            }
            char d = s[left];
            left++;
            if(need.count(d)){
                if(window[d] == need[d]) valid--;
                window[d]--;
            }
        }
    }
    return len == INT_MAX ? "" : s.substr(start, len);
}

int main(){
    string s = "ADOBECODEBANC";
    string t = "ABC";
    string ans = minWindow(s, t);
    cout << ans << "\n";
    return 0;
}
``

## 普通数组

### 最大子数组和
给你一个整数数组 nums ，请你找出一个具有最大和的连续子数组（子数组最少包含一个元素），返回其最大和。

子数组是数组中的一个连续部分。

思路：前缀和，同时记录之前的最小值。

实现：
```cpp
class Solution{
public:
    int maxSubArray(vector<int>& nums){
        int ans = INT_MIN;
        int s = 0;
        int s_min = 0;
        for(int num : nums){
            s += num;
            ans = max(ans, s - s_min);
            s_min = min(s_min, s);

        }
        return ans;
    }
}
```

### 合并区间
以数组 intervals 表示若干个区间的集合，其中单个区间为 intervals[i] = [starti, endi] 。请你合并所有重叠的区间，并返回 一个不重叠的区间数组，该数组需恰好覆盖输入中的所有区间 。

思路：先将区间按照起始位置排序，然后遍历排序后的区间列表，如果当前区间的起始位置在上一个区间的结束位置之前，说明两个区间重叠，需要合并，合并是考虑[1, 10]和[2, 6]这种情况，需要取最大值；否则说明两个区间不重叠，可以直接将当前区间加入结果列表。

注意：直接使用`ans.back()[1] = max(ans.back()[1], intervals[i][1])`来更新上一个区间的结束位置，保证了合并后的区间能够覆盖所有重叠的区间，不需要额外变量记录ans的长度。

实现：
```cpp
class Solution {
public:
    vector<vector<int>> merge(vector<vector<int>>& intervals) {
        std::sort(intervals.begin(), intervals.end());
        vector<vector<int>> ans;
        for(int num : intervals){
            if(ans.back()[1] >= num[0]){
                ans.back()[1] = max(ans.back()[1], num[1]);
            } else {
                ans.push_back(num);
            }
        }
        return ans;
    }
}；
```

### 轮转数组
给定一个整数数组 nums，将数组中的元素向右轮转 k 个位置，其中 k 是非负数。

思路：反转数组。首先将整个数组反转，然后分别反转前 k 个元素和后 n-k 个元素。

注意：`reverse(start, end)`表示左闭右闭区间，即反转的范围是[start, end)，所以在反转前 k 个元素时，应该使用`reverse(nums.begin(), nums.begin() + k)`，同样在反转后 n-k 个元素时，应该使用`reverse(nums.begin() + k, nums.end())`。

实现：
```cpp
class Solution {
public:
    void rotate(vector<int>& nums, int k) {
        int n = nums.size();
        k = k % n;
        reverse(nums.begin(), nums.end());
        reverse(nums.begin(), nums.begin() + k);
        reverse(nums.begin() + k, nums.end());
    }
};
```

### 除自身以外的数组的乘积
给你一个整数数组 nums，返回 数组 answer ，其中 answer[i] 等于 nums 中除了 nums[i] 之外其余各元素的乘积 。

题目数据 保证 数组 nums之中任意元素的全部前缀元素和后缀的乘积都在  32 位 整数范围内。

请 不要使用除法，且在 O(n) 时间复杂度内完成此题。出于对空间复杂度分析的目的，输出数组 不被视为 额外空间。

思路：使用前缀积和后缀积。先计算每个元素右侧的乘积，然后再遍历数组时，用一个变量记录左侧的乘积，同时更新结果数组。优化空间复杂度，直接在结果数组中存储后缀积，避免使用额外的数组来存储前缀积。

实现：
```cpp
class Solution {
public:
    vector<int> productExceptSelf(vector<int>& nums) {
        int n = nums.size();
        vector<int> suf(n);
        suf[n - 1] = 1;
        for(int i = n - 2; i >= 0; i--){
            suf[i] = suf[i + 1] * nums[i + 1];
        }
        int pre = 1;
        for(int i = 0; i < n; i++){
            suf[i] *= pre;
            pre *= nums[i]; 
        }
        return suf;
    }
};
```

### 缺失的第一个正数
给你一个未排序的整数数组 nums ，请你找出其中没有出现的最小的正整数。

请你实现时间复杂度为 O(n) 并且只使用常数级别额外空间的解决方案。

思路：首先观察标准的数组，每一个正数都应该在它对应的下标位置上，例如数字1应该在下标0的位置，数字2应该在下标1的位置，以此类推。我们可以通过遍历数组，将每个正数放到它对应的下标位置上，如果当前数字是x，并且x在1到n之间，那么我们就将它放到下标x-1的位置上。这样一来，最终我们只需要再次遍历数组，找到第一个位置i，使得nums[i] != i + 1，那么i + 1就是缺失的最小正整数。

注意：
1. 假如大数交换到了数组的开头怎么办？需不需要回过头看看已交换过的元素？不需要，因为在后续的交换中，这些元素总会被正确地放置到它们对应的位置上。
2. 判断条件：while中一定得写成`nums[i] != nums[nums[i] - 1]`，而不是`nums[i] != i + 1`，因为前者可以避免死循环的情况，例如当数组中存在重复元素时，如果写成后者可能会导致死循环。
3. 数值溢出：最后判断缺失的最小正整数时，应该写成`nums[i] != i + 1`而不是`nums[i] - 1 != i`，因为nums[i]可能是INT_MIN。而while循环中可以写成`nums[i] != nums[nums[i] - 1]`因为`&&`的短路特性可以避免访问越界的情况（必须满足`nums[i] > 0 && nums[i] <= n`）。

实现：
```cpp
class Solution {
public:
    int firstMissingPositive(vector<int>& nums) {
        int n = nums.size();
        for(int i = 0; i < n; i++){
            while(nums[i] > 0 && nums[i] <= n && nums[i] != nums[nums[i] - 1]){
                int j = nums[i] - 1;
                swap(nums[i], nums[j]);
            }
        }
        for(int i = 0; i < n; i++){
            if(nums[i] != i + 1){
                return i + 1;
            }
        }
        return n + 1;
    }
}；
```

### 矩阵置零
给定一个 m x n 的矩阵，如果一个元素为 0 ，则将其所在行和列的所有元素都设为 0 。请使用 原地 算法。

思路：遍历得到所有为0的元素的位置，然后将这些位置所在的行和列都置为0。为了节省空间，可以使用矩阵的第一行和第一列来记录哪些行和列需要被置零。由于[0,0]位置既属于第一行又属于第一列，因此需要额外的变量来记录第一行和第一列是否需要置零。

实现：
```cpp
class Solution {
public:
    void setZeroes(vector<vector<int>>& matrix) {
        int m = matrix.size(), n = matrix[0].size();
        bool col0 = false;

        for(int i = 0; i < m; i++){
            if(matrix[i][0] == 0) col0 = true;
            for(int j = 1; j < n; j++){
                if(matrix[i][j] == 0){
                    matrix[i][0] = 0;
                    matrix[0][j] = 0;
                }
            }
        }
        for(int i = m - 1; i >= 0; i--){
            for(int j = n - 1; j > 0; j--){
                if(matrix[i][0] == 0 || matrix[0][j] == 0){
                    matrix[i][j] = 0;
                }
            }
            if(col0) matrix[i][0] = 0;     
        }
    }
};
```

### 螺旋矩阵
给你一个 m 行 n 列的矩阵 matrix ，请按照 顺时针螺旋顺序 ，返回矩阵中的所有元素。

思路：定义四个变量top、bottom、left、right分别表示当前螺旋的上下左右边界，然后按照顺时针的顺序依次遍历矩阵，更新边界，直到所有元素都被访问。

实现：
```cpp
class Solution {
public:
    vector<int> spiralOrder(vector<vector<int>>& matrix) {
        vector<int> ans;
        if(matrix.empty()) return ans;
        int u = 0;
        int d = matrix.size() - 1;
        int l = 0;
        int r = matrix[0].size() - 1;
        while(true){
            for(int i = l; i <= r; i++) ans.push_back(matrix[u][i]);
            if(++u > d) break;
            for(int i = u; i <= d; i++) ans.push_back(matrix[i][r]);
            if(--r < l) break;
            for(int i = r; i >= l; i--) ans.push_back(matrix[d][i]);
            if(--d < u) break;
            for(int i = d; i >= u; i--) ans.push_back(matrix[i][l]);
            if(++l > r) break;
        }
        return ans;
    }
};
```

### 旋转图像
给定一个 n × n 的二维矩阵 matrix 表示一个图像。请你将图像顺时针旋转 90 度。

你必须在 原地 旋转图像，这意味着你需要直接修改输入的二维矩阵。请不要 使用另一个矩阵来旋转图像。

思路：90°旋转 = 转置 + 镜像；`reverse()`的使用。

实现：
```cpp
class Solution {
public:
    void rotate(vector<vector<int>>& matrix) {
        int n = matrix.size();
        for(int i = 0; i < n; i++){
            for(int j = 0; j < i; j++){
                swap(matrix[i][j], matrix[j][i]);
            }
        }
        for(int i = 0; i < n; i++){
            reverse(matrix[i].begin(), matrix[i].end());
        }
    }
};
```

### 搜索二维矩阵Ⅱ
编写一个高效的算法来搜索 m x n 矩阵 matrix 中的一个目标值 target 。该矩阵具有以下特性：
- 每行的元素从左到右升序排列。
- 每列的元素从上到下升序排列。

思路：利用矩阵的升序和降序性质，从右上角开始查找，如果当前元素大于 target，则向左移动，如果当前元素小于 target，则向下移动。（左下角开始也可以，但是左上角和右下角不行，因为两个方向是同样的递进关系）

实现：
```cpp
class Solution {
public:
    bool searchMatrix(vector<vector<int>>& matrix, int target) {
        int m = matrix.size(), n = matrix[0].size();
        int i = 0;
        int j = n - 1;
        while(i < m && j >= 0){
            if(matrix[i][j] > target){
                j--;
            }else if(matrix[i][j] < target){
                i++;
            }else return true;
        }
        return false;
    }
};
```

### 相交链表
给你两个单链表的头节点 headA 和 headB ，请你找出并返回两个单链表相交的起始节点。如果两个链表没有交点，返回 null 。

思路：使用双指针法。初始化两个指针 p 和 q，分别指向 headA 和 headB。然后同时移动两个指针，当指针到达链表末尾时，跳到另一个链表的头部继续移动。这样，当两个链表长度不同时，指针会在第二次遍历时对齐。如果存在相交节点，两个指针最终会在相交节点相遇；如果不存在相交节点，两个指针最终都会指向 null。

实现：
```cpp
#include <iostream>
#include <vector>
using namespace std;

struct ListNode{
    int val;
    ListNode* next;
    ListNode(int x) : val(x), next(nullptr) {}
};

ListNode* getIntersectionNode(ListNode* headA, ListNode* headB){
    ListNode* p = headA, *q = headB;
    while(p != q){
        p = p ? p->next : headB;//若p不为空，则p指向下一个节点；否则，p指向headB
        q = q ? q->next : headA;//若q不为空，则q指向下一个节点；否则，q指向headA
    }
    return p;
}

int main(){
    ListNode* common = new ListNode(8);
    common -> next = new ListNode(4);
    common -> next -> next = new ListNode(5);

    //A前半
    ListNode* headA = new ListNode(4);
    headA -> next = new ListNode(1);
    headA -> next -> next = common;

    //B前半
    ListNode* headB = new ListNode(5);
    headB -> next = new ListNode(6);
    headB -> next -> next = new ListNode(1);
    headB -> next -> next -> next = common;

    ListNode* ans = getIntersectionNode(headA, headB);
    cout << "Intersected at '" << ans -> val << "'\n";
    return 0;
}
```

### 反转链表
给你单链表的头节点 head ，请你反转链表，并返回反转后的链表的头节点。

思路：使用三个指针 prev、curr 和 next 来反转链表。初始化 prev 为 null，curr 为 head。然后遍历链表，在每一步中，先保存 curr 的下一个节点 next，然后将 curr 的 next 指向 prev，最后将 prev 和 curr 分别向后移动一位。

实现：
```cpp
#include <iostream>
using namespace std;

struct ListNode{
    int val;
    ListNode* next;
    ListNode(int x) : val(x), next(nullptr) {}
};

ListNode* reverseList(ListNode* head){
    ListNode* pre = nullptr;
    ListNode* cur = head;
    while(cur){
        ListNode* nxt = cur->next;
        cur->next = pre;
        pre = cur;
        cur = nxt;
    }
    return pre;
}

void printList(ListNode* head){
    while(head){
        cout << head->val << " ";
        head = head->next;
    }
    cout << "\n";
}

int main(){
    ListNode* n1 = new ListNode(1);
    ListNode* n2 = new ListNode(2);
    ListNode* n3 = new ListNode(3);
    ListNode* n4 = new ListNode(4);
    ListNode* n5 = new ListNode(5);
    n1->next = n2;
    n2->next = n3;
    n3->next = n4;
    n4->next = n5;

    ListNode* ans = reverseList(n1);
    printList(ans);
    return 0;
}
```

### 回文链表
给你一个单链表的头节点 head ，请你判断该链表是否为回文链表。如果是，返回 true ；否则，返回 false 。

思路：先找到链表的中点，然后反转后半部分链表，最后比较前半部分和后半部分的节点值是否相同。

实现：
```cpp
#include <iostream>
using namespace std;
struct ListNode {
    int val;
    ListNode *next;
    ListNode(int x) : val(x), next(NULL) {}
};

ListNode* reverseList(ListNode* head) {
    ListNode* pre = nullptr, *cur = head;
    while (cur) {
        ListNode* nxt = cur->next;
        cur->next = pre;
        pre = cur;
        cur = nxt;
    }
    return pre;
}

ListNode* middleNode(ListNode* head) {
    ListNode* slow = head, *fast = head;
    while (fast && fast->next) {
        slow = slow->next;
        fast = fast->next->next;
    }
    return slow;
}

bool isPalindrome(ListNode* head){
    ListNode* mid = middleNode(head);
    ListNode* rev = reverseList(mid);//反转前半段
    while(rev){
        if(rev->val != head->val) return false;
        rev = rev->next;
        head = head->next;
    }
    return true;
}

int main(){
    ListNode* head = new ListNode(1);
    head->next = new ListNode(2);
    head->next->next = new ListNode(2);
    head->next->next->next = new ListNode(1);

    cout << (isPalindrome(head) ? "true" : "false")  << "\n";
    return 0;
}
```

### 环形链表
给你一个链表的头节点 head ，判断链表中是否有环。如果链表中存在环，则返回 true 。 否则，返回 false 。

思路：快慢指针，快指针每次移动两步，慢指针每次移动一步，如果链表中存在环，那么快指针最终会追上慢指针；如果链表中不存在环，快指针会先到达链表末尾。

实现：
```cpp
#include <iostream>
#include <vector>
using namespace std;
struct ListNode {
    int val;
    ListNode *next;
    ListNode(int x) : val(x), next(NULL) {}
};

bool hasCycle(ListNode *head) {
    if(!head) return false;
    ListNode* slow = head, *fast = head;
    while(fast && fast->next){
        slow = slow->next;
        fast = fast->next->next;
        if(slow == fast) return true;
    }
    return false;
}

ListNode* bulidList(const vector<int>& nums){
    ListNode dummy(0);
    ListNode* tail = &dummy;
    for(int num : nums){
        tail->next = new ListNode(num);
        tail = tail->next;
    }
    return dummy.next;
}

int main(){
    ListNode* head = bulidList({3, 2, 0, -4});
    head->next->next->next->next = head->next; // Create a cycle
    cout << (hasCycle(head) ? "true" : "false") << "\n";
    return 0;
}
```

### 环形链表Ⅱ
给定一个链表，返回链表开始入环的第一个节点。 如果链表无环，则返回 null。

思路：快慢指针，当快慢指针相遇时，快指针回到链表头部，慢指针保持在相遇位置，然后两个指针每次移动一步，当它们再次相遇时，相遇点就是链表入环的第一个节点。

实现：
```cpp
#include <iostream>
#include <vector>
using namespace std;
struct ListNode {
    int val;
    ListNode *next;
    ListNode(int x) : val(x), next(NULL) {}
};

ListNode* detectCycle(ListNode *head) {
    if(!head) return nullptr;
    ListNode* slow = head, *fast = head;
    while(fast && fast->next){
        slow = slow->next;
        fast = fast->next->next;
        if(slow == fast){
            fast = head;
            while(slow != fast){
                slow = slow->next;
                fast = fast->next;
            }
            return slow;
        }
    }
    return nullptr;
}

ListNode* bulidList(const vector<int>& nums){
    ListNode dummy(0);
    ListNode* tail = &dummy;
    for(int num : nums){
        tail->next = new ListNode(num);
        tail = tail->next;
    }
    return dummy.next;
}

int main(){
    ListNode* head = bulidList({3, 2, 0, -4});
    head->next->next->next->next = head->next; // Create a cycle
    cout << detectCycle(head)->val << "\n";
    return 0;
}
```

### 合并两个有序链表
将两个升序链表合并为一个新的 升序 链表并返回。新链表是通过拼接给定的两个链表的所有节点组成的。

思路：每次都比较，将较小的节点接在结果链表的后面，然后移动对应链表的指针，直到其中一个链表遍历完，最后将另一个链表剩余的部分接在结果链表的后面。

注意：dummy哨兵节点的使用；l1和l2长度不一样时，剩余部分直接接在结果链表后面。

实现：
```cpp
#include <vector>
#include <iostream>
using namespace std;

struct ListNode {
    int val;
    ListNode *next;
    ListNode(int x) : val(x), next(NULL) {}
};

ListNode* mergeTwoLists(ListNode* l1, ListNode* l2) {
    if(!l1) return l2;
    if(!l2) return l1;
    ListNode dummy(0);
    ListNode* ans = &dummy;
    while(l1 && l2){
        if(l1->val > l2->val){
            ans->next = l2;
            l2 = l2->next;
        }
        else{
            ans->next = l1;
            l1 = l1->next;
        }
        ans = ans->next;
    }
    ans->next = l1 ? l1 : l2;//如果l1和l2长度不一样，剩下的部分直接接在后面
    return dummy.next;
}

ListNode* buildList(const vector<int>& nums){
    ListNode dummy(0);
    ListNode* head = &dummy;
    for(int num : nums){
        head->next = new ListNode(num);
        head = head->next;
    }
    return dummy.next;
}

int main(){
    vector<int> nums1 = {1, 2, 4};
    vector<int> nums2 = {1, 3, 4};
    ListNode* l1 = buildList(nums1);
    ListNode* l2 = buildList(nums2);
    ListNode* ans = mergeTwoLists(l1, l2);
    while(ans){
        cout << ans->val << " ";
        ans = ans->next;
    }
    cout << "\n";
    return 0;
}
```

### 两数相加
给你两个非空的链表，表示两个非负的整数。它们每位数字都是按照逆序的方式存储的，并且每个节点只能存储一位数字。请你将两个数相加，并以相同形式返回一个表示和的链表。

思路：用carry和sum记录进位和当前位的和，遍历两个链表，直到两个链表都遍历完，并且carry为0。每次计算sum = carry + (l1 ? l1->val : 0) + (l2 ? l2->val : 0)，然后更新carry = sum / 10，并将sum % 10作为当前位的值加入结果链表。

实现：
```cpp
#include <vector>
#include <iostream>
using namespace std;

struct ListNode {
    int val;
    ListNode *next;
    ListNode(int x) : val(x), next(NULL) {}
};

ListNode* buildList(const vector<int>& nums){
    ListNode dummy(0);
    ListNode* head = &dummy;
    for(int num : nums){
        head->next = new ListNode(num);
        head = head->next;
    }
    return dummy.next;
}

ListNode* addTwoNumbers(ListNode* l1, ListNode* l2) {
    ListNode dummy(0);
    ListNode* head = &dummy;
    int carry = 0;
    while(l1 || l2 || carry){
        int sum = carry;//记录进位
        if(l1){
            sum += l1->val;
            l1 = l1->next;
        }
        if(l2){
            sum += l2->val;
            l2 = l2->next;
        }
        carry = sum / 10;
        head->next = new ListNode(sum % 10);
        head = head->next;
    }
    return dummy.next;
}

int main(){
    vector<int> nums1 = {2, 4, 3};
    vector<int> nums2 = {5, 6, 4};
    ListNode* l1 = buildList(nums1);
    ListNode* l2 = buildList(nums2);
    ListNode* ans = addTwoNumbers(l1, l2);
    while(ans){
        cout << ans->val << " ";
        ans = ans->next;
    }
    cout << "\n";
    return 0;
}
```

### 删除链表的倒数第N个节点
给你一个链表，删除链表的倒数第 n 个节点，并且返回链表的头结点。

思路：用left和right两个指针，right先移动n步，然后left和right同时移动，直到right到达链表末尾，此时left指向的就是倒数第n个节点的前一个节点，然后将left->next指向left->next->next，即可删除倒数第n个节点。

实现：
```cpp
#include <vector>
#include <iostream>
using namespace std;

struct ListNode {
    int val;
    ListNode *next;
    ListNode(int x) : val(x), next(NULL) {}
};

ListNode* buildList(const vector<int>& nums){
    ListNode dummy(0);
    ListNode* head = &dummy;
    for(int num : nums){
        head->next = new ListNode(num);
        head = head->next;
    }
    return dummy.next;
}

ListNode* removeNthFromEnd(ListNode* head, int n) {
    ListNode dummy(0);
    dummy.next = head;
    ListNode* left = &dummy;
    ListNode* right = &dummy;
    while(n--) right = right->next;
    while(right->next){
        left = left->next;
        right = right->next;
    }
    ListNode* toDelete = left->next;
    left->next = left->next->next;
    delete toDelete;
    return dummy.next;
}

int main(){
    vector<int> nums = {1, 2, 3, 4, 5};
    int n = 2;
    ListNode* head = buildList(nums);
    head = removeNthFromEnd(head, n);
    while(head){
        cout << head->val << " ";
        head = head->next;
    }
    cout << "\n";
    return 0;
}
```

### 两两交换链表中的节点
给你一个链表，两两交换其中相邻的节点，并返回交换后链表的头节点。

思路：定义四个指针，分别为pre、cur、nxt和nxtnxt，pre指向当前要交换的两个节点的前一个节点，cur和nxt分别指向当前要交换的两个节点，nxtnxt指向当前要交换的两个节点的下一个节点。交换后，pre->next指向nxt，cur->next指向nxtnxt，nxt->next指向cur，然后将pre移动到cur的位置，将cur移动到nxtnxt的位置，继续下一轮交换。

实现：
```cpp
#include <vector>
#include <iostream>
using namespace std;

struct ListNode {
    int val;
    ListNode *next;
    ListNode(int x) : val(x), next(NULL) {}
};

ListNode* buildList(const vector<int>& nums){
    ListNode dummy(0);
    ListNode* head = &dummy;
    for(int num : nums){
        head->next = new ListNode(num);
        head = head->next;
    }
    return dummy.next;
}

ListNode* swapPairs(ListNode* head){
    ListNode dummy(0);
    dummy.next = head;
    ListNode* pre = &dummy;
    ListNode* cur = head;
    while(cur && cur->next){
        ListNode* nxt = cur->next;
        ListNode* nxtnxt = cur->next->next;

        pre->next = nxt;
        nxt->next = cur;
        cur->next = nxtnxt;

        pre = cur; // 更新pre，注意是cur
        cur = nxtnxt;
    }
    return dummy.next;
}

int main(){
    vector<int> nums = {1, 2, 3, 4};
    ListNode* head = buildList(nums);
    ListNode* ans = swapPairs(head);
    while(ans){
        cout << ans->val << " ";
        ans = ans->next;
    }
    cout << "\n";
    return 0;
}
```

### k个一组翻转链表
给你一个链表，每 k 个节点一组进行翻转，请你返回翻转后的链表。

思路：定义四个指针，分别为p0、pre、cur和nxt，p0指向当前要翻转的k个节点的前一个节点，pre指向当前要翻转的k个节点的前一个节点，cur指向当前要翻转的k个节点的第一个节点，nxt指向当前要翻转的k个节点的下一个节点。翻转后，p0->next指向pre，cur->next指向nxt，然后将p0移动到cur的位置，将cur移动到nxt的位置，继续下一轮翻转。

实现：
```cpp
#include <vector>
#include <iostream>
using namespace std;

struct ListNode {
    int val;
    ListNode *next;
    ListNode(int x) : val(x), next(NULL) {}
};

ListNode* buildList(const vector<int>& nums){
    ListNode dummy(0);
    ListNode* head = &dummy;
    for(int num : nums){
        head->next = new ListNode(num);
        head = head->next;
    }
    return dummy.next;
}

ListNode* reverseKGroup(ListNode* head, int k){
    int n = 0;
    for(ListNode* cur = head; cur; cur = cur->next) n++;
    ListNode dummy(0);
    dummy.next = head;
    ListNode* p0 = &dummy;
    ListNode* pre = nullptr;
    ListNode* cur = head;
    for(; n >= k; n -= k){
        for(int i = 0; i < k; i++){
            ListNode* nxt = cur->next;
            cur->next = pre;
            pre = cur;
            cur = nxt;
        }
        ListNode* nxt = p0->next;//保存当前段的头结点
        p0->next->next = cur;//当前段的头结点指向下一段的头结点
        p0->next = pre;//当前段的头结点指向反转后的头结点
        p0 = nxt;//移动p0到下一段的前一个节点
    }
    return dummy.next;
    //以[1,2,3,4]且k=3为例：
    //初始化：p0指向dummy，pre=nullptr，cur=1  dummy->1->2->3->4->nullptr
    //第一次循环：1指向nullptr，pre=1，cur=2    1->nullptr,2->3->4->nullptr
    //第二次循环：2指向1，pre=2，cur=3          2->1->nullptr,3->4->nullptr
    //第三次循环：3指向2，pre=3，cur=4          3->2->1->nullptr,4->nullptr
    //nxt = p0->next指向1                     dummy->1，保留1
    //将1指向4                                 1->4->nullptr  
    //将p0->next指向3                          dummy->3->2->1->4->nullptr
    //将p0移动到1                              p0更新为1，准备下一段的反转
}

int main(){
    vector<int> nums = {1, 2, 3, 4, 5, 6, 7};
    ListNode* head = buildList(nums);
    ListNode* ans = reverseKGroup(head, 3);
    while(ans){
        cout << ans->val << " ";
        ans = ans->next;
    }
    cout << "\n";
    return 0;
}
```

### 随机链表的复制
给你一个长度为 n 的链表，每个节点包含一个额外增加的随机指针 random ，该指针可以指向链表中的任何节点或空节点。

构造这个链表的 深拷贝。 深拷贝应该正好由 n 个 全新 节点组成，其中每个新节点的值都设为其对应的原节点的值。新节点的 next 指针和 random 指针也都应指向复制链表中的新节点，并使原链表和复制链表中的这些指针能够表示相同的链表状态。复制链表中的指针都不应指向原链表中的节点 。

例如，如果原链表中有 X 和 Y 两个节点，其中 X.random --> Y 。那么在复制链表中对应的两个节点 x 和 y ，同样有 x.random --> y 。

返回复制链表的头节点。

用一个由 n 个节点组成的链表来表示输入/输出中的链表。每个节点用一个 [val, random_index] 表示：

val：一个表示 Node.val 的整数。
random_index：随机指针指向的节点索引（范围从 0 到 n-1）；如果不指向任何节点，则为  null 。
你的代码 只 接受原链表的头节点 head 作为传入参数。

思路：在原链表的每个节点后复制粘贴一个新节点，然后再处理随机指针，最后将新旧节点分离。

实现：
```cpp
#include <vector>
#include <iostream>
using namespace std;

class Node {
public:
    int val;
    Node* next;
    Node* random;

    Node(int _val){
        val = _val;
        next = nullptr;
        random = nullptr;
    }
};

Node* copyRandomList(Node* head){
    if(!head) return nullptr;
    //核心是在原链表的每个节点后面复制粘贴一个新节点
    //第一步，复制，不拷贝随机节点
    for(Node* cur = head; cur; cur = cur->next->next){
        Node* next = cur->next;
        cur->next = new Node(cur->val); //创建新节点
        cur->next->next = next;         //复原指向
    }
    //将新节点的random对应过去
    for(Node* cur = head; cur; cur = cur->next->next){
        if(cur->random) cur->next->random = cur->random->next;
        //默认指向nullptr，无需处理
    }
    Node dummy(0);
    Node* tail = &dummy;
    for(Node* cur = head; cur; ){
        Node* copy = cur->next;
        tail->next = copy;
        cur->next = copy->next;
        cur = cur->next;//相当于cur = cur->next->next
        tail = tail->next;
    }
    return dummy.next;
}

int main(){
    Node* head = new Node(1);
    head->next = new Node(2);
    head->next->next = new Node(3);
    head->random = head->next;
    head->next->random = head->next;
    head->next->next->random = head;
    
    Node* ans = copyRandomList(head);
    cout << ans->val << " " << ans->next->val << endl;
    cout << ans->random->val << " " << ans->next->random->val << endl;
    return 0;
}
```

### 排序链表
给你链表的头结点 head ，请将其按 升序 排列并返回 排序后的链表 。

思路：使用归并排序，先找到链表的中点，将链表分成两半，然后递归地对两半进行排序，最后将两半合并。

实现：
```cpp
#include <vector>
#include <iostream>
using namespace std;

struct ListNode {
    int val;
    ListNode* next;
    ListNode(int x) : val(x), next(nullptr) {}
};

ListNode* buildList(const vector<int>& nums){
    ListNode dummy(0);
    ListNode* head = &dummy;
    for(int num : nums){
        head->next = new ListNode(num);
        head = head->next;
    }
    return dummy.next;
}

ListNode* middleNode(ListNode* head) {
    ListNode* pre = head;
    ListNode* slow = head;
    ListNode* fast = head;
    while(fast && fast->next){
        pre = slow;
        slow = slow->next;
        fast = fast->next->next;
    }
    pre->next = nullptr;
    return slow;
}

ListNode* mergeTwoLists(ListNode* l1, ListNode* l2){
    ListNode dummy(0);
    ListNode* cur = &dummy;
    while(l1 && l2){
        if(l1->val < l2->val){
            cur->next = l1;
            l1 = l1->next;
        }else{
            cur->next = l2;
            l2 = l2->next;
        }
        cur = cur->next;
    }
    cur->next = l1 ? l1 : l2;
    return dummy.next;
}

ListNode* sortList(ListNode* head){
    if(head == nullptr || head->next == nullptr) return head;
    ListNode* head2 = middleNode(head);
    head = sortList(head);
    head2 = sortList(head2);
    return mergeTwoLists(head, head2);
}

int main(){
    vector<int> nums = {-1, 5, 3, 4, 0};
    ListNode* head = buildList(nums);
    ListNode* ans = sortList(head);
    for(ListNode* cur = ans; cur; cur = cur->next){
        cout << cur->val << " ";
    }
    cout << endl;
    return 0;
}
```

### 合并K个升序链表
给你一个链表数组，每个链表都已经按升序排列。请你将所有链表合并到一个升序链表中，返回合并后的链表。

思路：使用分治法，先将链表数组两两合并，得到一个新的链表数组，然后继续两两合并，直到只剩下一个链表。

实现：
```cpp
#include <vector>
#include <iostream>
using namespace std;

struct ListNode {
    int val;
    ListNode* next;
    ListNode(int x) : val(x), next(nullptr) {}
};

ListNode* buildList(const vector<int>& nums){
    ListNode dummy(0);
    ListNode* head = &dummy;
    for(int num : nums){
        head->next = new ListNode(num);
        head = head->next;
    }
    return dummy.next;
}

ListNode* mergeTwoLists(ListNode* l1, ListNode* l2){
    ListNode dummy(0);
    ListNode* cur = &dummy;
    while(l1 && l2){
        if(l1->val < l2->val){
            cur->next = l1;
            l1 = l1->next;
        }else{
            cur->next = l2;
            l2 = l2->next;
        }
        cur = cur->next;
    }
    cur->next = l1 ? l1 : l2;
    return dummy.next;
}

ListNode* mergeKLists(vector<ListNode*>& lists, int i, int j){
    int m = j - i;
    if(m == 0) return nullptr;
    if(m == 1) return lists[i];
    auto left = mergeKLists(lists, i, i + m / 2);
    auto right = mergeKLists(lists, i + m / 2, j);
    return mergeTwoLists(left, right);
}

int main(){
    vector<ListNode*> lists;
    lists.push_back(buildList({1, 4, 5}));
    lists.push_back(buildList({1, 3, 4}));
    lists.push_back(buildList({2, 6}));
    
    ListNode* ans = mergeKLists(lists, 0, lists.size());
    for(ListNode* cur = ans; cur; cur = cur->next){
        cout << cur->val << " ";
    }
    cout << endl;
    return 0;
}
```

### LRU缓存
实现：
```cpp
#include <vector>
#include <iostream>
#include <unordered_map>
using namespace std;

struct Node{
    int key;
    int value;
    Node* prev;
    Node* next;

    Node(int k = 0, int v = 0) : key(k), value(v) {}
};

class LRUCache{
private:
    int capacity;
    Node* dummy;
    unordered_map<int, Node*> key_to_node;

    void remove(Node* x){
        x->prev->next = x->next;
        x->next->prev = x->prev;
    }

    void push_front(Node* x){
        x->prev = dummy;
        x->next = dummy->next;
        x->prev->next = x;
        x->next->prev = x;
    }

    Node* get_node(int key){
        if(!key_to_node.count(key)) return nullptr;
        Node* node = key_to_node[key];//哈希表的访问会把这个节点提升到最近使用的位置
        remove(node);
        push_front(node);
        return node;
    }

public:
    LRUCache(int capacity): capacity(capacity), dummy(new Node()){
        dummy->next = dummy;
        dummy->prev = dummy;
    }
    int get(int key){
        Node* node = get_node(key);
        return node ? node->value : -1;
    }
    void put(int key, int value){
        Node* node = get_node(key);
        if(node){
            node->value = value;
            return;
        }
        key_to_node[key] = node = new Node(key, value);
        push_front(node);
        if(key_to_node.size() > capacity){
            Node* last = dummy->prev;
            key_to_node.erase(last->key);
            remove(last);
            delete last;
        } 
    }
};

int main(){
    LRUCache cache(2);
    cache.put(1, 1);
    cache.put(2, 2);
    cout << cache.get(1) << endl;   // 返回 1
    cache.put(3, 3);                // 淘汰 2
    cout << cache.get(2) << endl;   // 返回 -1 (未找到)
    cache.put(4, 4);                // 淘汰 1
    cout << cache.get(1) << endl;   // 返回 -1 (未找到)
    cout << cache.get(3) << endl;   // 返回 3
    cout << cache.get(4) << endl;   // 返回 4
    return 0;
}
```

### 二叉树的中序遍历
给定一个二叉树的根节点 root ，返回它的 中序 遍历。

思路：递归法，先遍历左子树，再访问根节点，最后遍历右子树。

实现：
```cpp
#include <iostream>
#include <queue>
#include <vector>

using namespace std;

struct TreeNode {
	int val;
	TreeNode* left;
	TreeNode* right;

	TreeNode(int x) : val(x), left(nullptr), right(nullptr) {}
};

TreeNode* buildTreeNode(const vector<string>& nodes) {
	if (nodes.empty() || nodes[0] == "null") {
		return nullptr;
	}

	TreeNode* root = new TreeNode(stoi(nodes[0]));
	queue<TreeNode*> nodeQueue;
	nodeQueue.push(root);
	int index = 1;

	while (!nodeQueue.empty() && index < static_cast<int>(nodes.size())) {
		TreeNode* current = nodeQueue.front();
		nodeQueue.pop();

		if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
			current->left = new TreeNode(stoi(nodes[index]));
			nodeQueue.push(current->left);
		}
		++index;

		if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
			current->right = new TreeNode(stoi(nodes[index]));
			nodeQueue.push(current->right);
		}
		++index;
	}

	return root;
}

vector<int> dfs(TreeNode* root, vector<int>& result) {
	if (!root) {
		return result;
	}
	dfs(root->left, result);
	result.push_back(root->val);
	dfs(root->right, result);
	return result;
}

int main() {
    vector<string> nodes = {"1", "null", "2", "3"};
    TreeNode* root = buildTreeNode(nodes);
    vector<int> result;
    result = dfs(root, result);

    for (int val : result) {
        cout << val << " ";
    }
    cout << "\n";

    return 0;
}
```

### 二叉树的最大深度
给定一个二叉树，找出其最大深度。二叉树的深度为根节点到最远叶子节点的最长路径上的节点数。

思路：递归法，最大深度等于左子树和右子树的最大深度的较大值加1。

实现：
```cpp
#include <iostream>
#include <queue>
#include <vector>
using namespace std;

struct TreeNode{
    int val;
    TreeNode* left;
    TreeNode* right;

    TreeNode(int x) : val(x), left(nullptr), right(nullptr) {}
};

TreeNode* buildTreeNode(const vector<string>& nodes) {
	if (nodes.empty() || nodes[0] == "null") {
		return nullptr;
	}

	TreeNode* root = new TreeNode(stoi(nodes[0]));
	queue<TreeNode*> nodeQueue;
	nodeQueue.push(root);
	int index = 1;

	while (!nodeQueue.empty() && index < static_cast<int>(nodes.size())) {
		TreeNode* current = nodeQueue.front();
		nodeQueue.pop();

		if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
			current->left = new TreeNode(stoi(nodes[index]));
			nodeQueue.push(current->left);
		}
		++index;

		if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
			current->right = new TreeNode(stoi(nodes[index]));
			nodeQueue.push(current->right);
		}
		++index;
	}

	return root;
}

int maxDepth(TreeNode* root){
    if(!root) return 0;
    int l = maxDepth(root->left);
    int r = maxDepth(root->right);
    return max(l, r) + 1;
}

int main(){
    vector<string> nodes = {"3", "9", "20", "null", "null", "15", "7"};
    TreeNode* root = buildTreeNode(nodes);
    cout << maxDepth(root) << "\n";
    return 0;
}
```

```cpp
#include <iostream>
#include <queue>
#include <vector>
#include <algorithm>
using namespace std;

struct TreeNode {
    int val;
    TreeNode* left;
    TreeNode* right;

    TreeNode(int x) : val(x), left(nullptr), right(nullptr) {}
};

TreeNode* buildTreeNode(const vector<string>& nodes) {
	if (nodes.empty() || nodes[0] == "null") {
		return nullptr;
	}

	TreeNode* root = new TreeNode(stoi(nodes[0]));
	queue<TreeNode*> nodeQueue;
	nodeQueue.push(root);
	int index = 1;

	while (!nodeQueue.empty() && index < static_cast<int>(nodes.size())) {
		TreeNode* current = nodeQueue.front();
		nodeQueue.pop();

		if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
			current->left = new TreeNode(stoi(nodes[index]));
			nodeQueue.push(current->left);
		}
		++index;

		if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
			current->right = new TreeNode(stoi(nodes[index]));
			nodeQueue.push(current->right);
		}
		++index;
	}

	return root;
}

void printTree(TreeNode* root){
    if(!root) return;
    cout << root->val << " ";
    printTree(root->left);
    printTree(root->right);
}//前序遍历打印树

TreeNode* invertTree(TreeNode* root){
    if(!root) return nullptr;
    swap(root->left, root->right);
    invertTree(root->left);
    invertTree(root->right);
    return root;
}

int main(){
    vector<string> nodes = {"4", "2", "7", "1", "3", "6", "9"};
    TreeNode* root = buildTreeNode(nodes);
    root = invertTree(root);
    printTree(root);
    cout << endl;
    return 0;
}
```

### 对称二叉树

### 二叉树的直径
给定一棵二叉树，你需要计算它的直径长度。二叉树的直径长度是树中任意两个节点路径长度中的最大值。这条路径可能穿过也可能不穿过根节点。

思路：递归法。观察边的数量，若节点为空，返回-1；否则定义l为左侧子树当前节点的直径，r为右侧子树，每次递归左右子树的值+1，更新ans为max(ans, l + r)；更新全局变量diameter为当前直径和之前直径的较大值，然后返回当前节点的高度，即max(l, r) 。

实现：
```cpp
#include <iostream>
#include <queue>
#include <vector>
#include <algorithm>
using namespace std;

struct TreeNode {
    int val;
    TreeNode* left;
    TreeNode* right;

    TreeNode(int x) : val(x), left(nullptr), right(nullptr) {}
};

TreeNode* buildTreeNode(const vector<string>& nodes) {
	if (nodes.empty() || nodes[0] == "null") {
		return nullptr;
	}

	TreeNode* root = new TreeNode(stoi(nodes[0]));
	queue<TreeNode*> nodeQueue;
	nodeQueue.push(root);
	int index = 1;

	while (!nodeQueue.empty() && index < static_cast<int>(nodes.size())) {
		TreeNode* current = nodeQueue.front();
		nodeQueue.pop();

		if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
			current->left = new TreeNode(stoi(nodes[index]));
			nodeQueue.push(current->left);
		}
		++index;

		if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
			current->right = new TreeNode(stoi(nodes[index]));
			nodeQueue.push(current->right);
		}
		++index;
	}

	return root;
}

int dfs(TreeNode* root, int& ans){
    if(root == nullptr) return -1;// 以边的数量为单位，所以空节点返回-1
    int l = dfs(root->left, ans) + 1;   // 以边的数量为单位，所以要加1
    int r = dfs(root->right, ans) + 1;  // 以边的数量为单位，所以要加1
    ans = max(ans, l + r);
    return max(l, r);// 返回以当前节点为根的子树的最大深度（以边的数量为单位）
}

int diameterOfBinaryTree(TreeNode* root) {
    int ans = 0;
    dfs(root, ans);
    return ans;
}

int main() {
    vector<string> nodes = {"1", "2", "3", "4", "5"};
    TreeNode* root = buildTreeNode(nodes);
    int result = diameterOfBinaryTree(root);
    cout << result << endl;
    return 0;
}
```

### 二叉树的层序遍历
给你一个二叉树，请你返回其按 层序遍历 得到的节点值。 （即逐层地，从左到右访问所有节点）。

思路：用一个cur二叉树节点数组来存储当前层的节点，遍历cur数组，将每个节点的值加入当前层的结果vals数组中，并将其左右子节点加入下一个层的节点数组next中。最后将vals加入结果ans中，并将cur更新为next，继续下一层的遍历。

注意：`std::move`的使用，可以避免不必要的复制，提高效率。

实现：
```cpp
#include <vector>
#include <queue>
#include <string>
#include <iostream>
using namespace std;

struct TreeNode {
    int val;
    TreeNode* left;
    TreeNode* right;

    TreeNode(int x) : val(x), left(nullptr), right(nullptr) {}
};

TreeNode* buildTreeNode(const vector<string>& nodes) {
    if (nodes.empty() || nodes[0] == "null") {
        return nullptr;
    }

    TreeNode* root = new TreeNode(stoi(nodes[0]));
    queue<TreeNode*> nodeQueue;
    nodeQueue.push(root);
    int index = 1;

    while (!nodeQueue.empty() && index < static_cast<int>(nodes.size())) {
        TreeNode* current = nodeQueue.front();
        nodeQueue.pop();

        if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
            current->left = new TreeNode(stoi(nodes[index]));
            nodeQueue.push(current->left);
        }
        ++index;

        if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
            current->right = new TreeNode(stoi(nodes[index]));
            nodeQueue.push(current->right);
        }
        ++index;
    }

    return root;
}

vector<vector<int>> levelOrder(TreeNode* root) {
    if(!root) return {};
    vector<vector<int>> ans;
    vector<TreeNode*> cur = {root};
    while(cur.size()){
        vector<TreeNode*> next;
        vector<int> vals;
        for(TreeNode* node : cur){
            vals.push_back(node->val);
            if(node->left) next.push_back(node->left);
            if(node->right) next.push_back(node->right);
        }
        cur = move(next);
        ans.push_back(move(vals));
    }
    return ans;
}

int main(){
    vector<string> nodes = {"3", "9", "20", "null", "null", "15", "7"};
    TreeNode* root = buildTreeNode(nodes);
    vector<vector<int>> result = levelOrder(root);

    for (const auto& level : result) {
        for (int val : level) {
            cout << val << " ";
        }
        cout << "\n";
    }
    return 0;
}
```

### 将有序数组转换成二叉搜索树
给你一个整数数组 nums ，其中元素已经按 升序 排列，请你将其转换为一棵高度平衡 二叉搜索树。

思路：递归法，找到数组中点然后返回一个以中点为根节点的树，左子树由数组左半部分构成，右子树由数组右半部分构成。

注意：TreeNode中需要定义一个构造函数来初始化节点的值和左右子树指针；使用left + (right - left) / 2来计算中点，避免整数溢出。

实现：
```cpp
#include <vector>
#include <queue>
#include <string>
#include <iostream>
using namespace std;

struct TreeNode {
    int val;
    TreeNode* left;
    TreeNode* right;

    TreeNode(int x) : val(x), left(nullptr), right(nullptr) {}
    TreeNode(int x, TreeNode* left, TreeNode* right) : val(x), left(left), right(right) {}
};

TreeNode* dfs(vector<int>& nums, int left, int right) {
    if(left == right) return nullptr;
    int m = left + (right - left) / 2;
    return new TreeNode(nums[m], dfs(nums, left, m), dfs(nums, m + 1, right));   
}

TreeNode* sortedArrayToBST(vector<int>& nums){
    return dfs(nums, 0, nums.size());
}

void printlevelOrder(TreeNode* root){
    if(!root){
        cout << "null" << endl;
        return;
    }

    queue<TreeNode*> q;
    vector<string> result;
    q.push(root);

    while(!q.empty()){
        TreeNode* node = q.front();
        q.pop();

        if(node == nullptr){
            result.push_back("null");
            continue;
        }

        result.push_back(to_string(node->val));
        q.push(node->left);
        q.push(node->right);
    }

    while(!result.empty() && result.back() == "null"){
        result.pop_back();
    }

    for(int i = 0; i < static_cast<int>(result.size()); ++i){
        if(i > 0) cout << " ";
        cout << result[i];
    }
    cout << endl;
}

int main() {
    vector<int> nums = {-10, -3, 0, 5, 9};
    TreeNode* root = sortedArrayToBST(nums);
    printlevelOrder(root);
    return 0;
}
```

### 验证二叉搜索树
给你一个二叉树的根节点 root ，判断其是否是一个有效的二叉搜索树。

思路：递归法。先设置一个最小值min = LLONG_MIN，遇到nullptr则返回true；如果当前节点的左子树不满足二叉搜索树，返回false；如果当前节点的值小于min，返回false；更新min为当前节点的值；返回当前节点的右子树。

实现：
```cpp
#include <vector>
#include <queue>
#include <string>
#include <iostream>
#include <climits>
using namespace std;

struct TreeNode {
    int val;
    TreeNode* left;
    TreeNode* right;

    TreeNode(int x) : val(x), left(nullptr), right(nullptr) {}
};

TreeNode* buildTreeNode(const vector<string>& nodes) {
    if (nodes.empty() || nodes[0] == "null") {
        return nullptr;
    }

    TreeNode* root = new TreeNode(stoi(nodes[0]));
    queue<TreeNode*> nodeQueue;
    nodeQueue.push(root);
    int index = 1;

    while (!nodeQueue.empty() && index < static_cast<int>(nodes.size())) {
        TreeNode* current = nodeQueue.front();
        nodeQueue.pop();

        if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
            current->left = new TreeNode(stoi(nodes[index]));
            nodeQueue.push(current->left);
        }
        ++index;

        if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
            current->right = new TreeNode(stoi(nodes[index]));
            nodeQueue.push(current->right);
        }
        ++index;
    }

    return root;
}

bool isValidBST(TreeNode* root, long long min = LLONG_MIN){
    if(root == nullptr) return true;
    if(!isValidBST(root->left, min)) return false;// 先判断左子树是否合法，如果不合法直接返回 false
    if(root->val <= min) return false;// 如果当前节点的值不大于前一个节点的值，说明不是二叉搜索树，返回 false
    min = root->val;
    return isValidBST(root->right, min);// 最后判断右子树是否合法，如果不合法直接返回 false，否则返回 true
}

int main() {
    vector<string> nodes = {"5", "1", "4", "null", "null", "3", "6"};
    TreeNode* root = buildTreeNode(nodes);
    cout << boolalpha << isValidBST(root) << endl;
    return 0;
}
```

### 二叉搜索树中第k小的元素
给定一个二叉搜索树，编写一个函数 kthSmallest 来查找其中第 k 个最小的元素。（从1开始计数）

思路：递归，先遍历左子树，并使k--，如果k为0，说明当前节点就是第k小的元素，返回当前节点的值；否则继续遍历右子树。

实现：
```cpp
#include <vector>
#include <queue>
#include <string>
#include <iostream>
#include <climits>
using namespace std;

struct TreeNode {
    int val;
    TreeNode* left;
    TreeNode* right;

    TreeNode(int x) : val(x), left(nullptr), right(nullptr) {}
};

TreeNode* buildTreeNode(const vector<string>& nodes) {
    if (nodes.empty() || nodes[0] == "null") {
        return nullptr;
    }

    TreeNode* root = new TreeNode(stoi(nodes[0]));
    queue<TreeNode*> nodeQueue;
    nodeQueue.push(root);
    int index = 1;

    while (!nodeQueue.empty() && index < static_cast<int>(nodes.size())) {
        TreeNode* current = nodeQueue.front();
        nodeQueue.pop();

        if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
            current->left = new TreeNode(stoi(nodes[index]));
            nodeQueue.push(current->left);
        }
        ++index;

        if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
            current->right = new TreeNode(stoi(nodes[index]));
            nodeQueue.push(current->right);
        }
        ++index;
    }

    return root;
}

void dfs(TreeNode* node, int& k, int& ans){
    if(node == nullptr || k <= 0) return;
    dfs(node->left, k, ans);
    k--;
    if(k == 0){
        ans = node->val;
        return;
    }
    dfs(node->right, k, ans);
}

int main(){
    vector<string> nodes = {"5", "3", "6", "2", "4", "null", "null", "1"};
    TreeNode* root = buildTreeNode(nodes);
    int k = 3;
    int ans = 0;
    dfs(root, k, ans);
    cout << ans << endl;
    return 0;
}
```

### 二叉树的右视图
给定一个二叉树，想象自己站在它的右侧，按照从顶部到底部的顺序，返回从右侧所能看到的节点值。

思路：递归，用一个depth表示当前深度，先访问右子树，这样在同一深度时右子树的节点会覆盖左子树的节点；判断depth是否等于ans.size()，只有第一次访问这个深度时才会添加节点值。

实现：
```cpp
#include <vector>
#include <queue>
#include <string>
#include <iostream>
#include <climits>
using namespace std;

struct TreeNode {
    int val;
    TreeNode* left;
    TreeNode* right;

    TreeNode(int x) : val(x), left(nullptr), right(nullptr) {}
};

TreeNode* buildTreeNode(const vector<string>& nodes) {
    if (nodes.empty() || nodes[0] == "null") {
        return nullptr;
    }

    TreeNode* root = new TreeNode(stoi(nodes[0]));
    queue<TreeNode*> nodeQueue;
    nodeQueue.push(root);
    int index = 1;

    while (!nodeQueue.empty() && index < static_cast<int>(nodes.size())) {
        TreeNode* current = nodeQueue.front();
        nodeQueue.pop();

        if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
            current->left = new TreeNode(stoi(nodes[index]));
            nodeQueue.push(current->left);
        }
        ++index;

        if (index < static_cast<int>(nodes.size()) && nodes[index] != "null") {
            current->right = new TreeNode(stoi(nodes[index]));
            nodeQueue.push(current->right);
        }
        ++index;
    }
    return root;
}

void dfs(TreeNode* node, int depth, vector<int>& ans){
    if(node == nullptr) return;
    if(depth == ans.size()) ans.push_back(node->val);
    // 只有第一次访问这个深度时才会添加节点值
    dfs(node->right, depth + 1, ans);
    // 先访问右子树，这样在同一深度时右子树的节点会覆盖左子树的节点
    dfs(node->left, depth + 1, ans);
}

int main(){
    vector<string> nodes = {"1", "2", "3", "null", "5", "null", "4"};
    TreeNode* root = buildTreeNode(nodes);
    vector<int> ans;
    dfs(root, 0, ans);
    for(int num : ans){
        cout << num << " ";
    }
    cout << "\n";
    return 0;
}
```

思路：递归。定义一个head节点，先递归地展开右子树，再递归地展开左子树，然后将当前节点的左子树置空，右子树指向head，最后更新head为当前节点。

实现：

### 有效的括号
给定一个只包括 '('，')'，'{'，'}'，'\['，'\]' 的字符串 s ，判断字符串是否有效。

有效字符串需满足：
- 左括号必须用相同类型的右括号闭合。
- 左括号必须以正确的顺序闭合。
- 每个右括号都有一个对应的相同类型的左括号。

思路：使用栈来匹配括号，遍历字符串，遇到左括号就入栈，遇到右括号就判断栈顶是否为对应的左括号，如果是则出栈，否则返回false，最后栈为空则字符串有效。

注意：满足条件的第二点说明左括号必须按照栈的先进后出顺序闭合，也就是说最近入栈的左括号必须最先被匹配的右括号闭合；且必须以'('、'{'、'\['这样的顺序。

实现：
```cpp
class Solution {
    unordered_map<char, char> mp = {{')', '('}, {']', '['}, {'}', '{'}};// 右括号为key
public:
    bool isValid(string s) {
        if(s.length() % 2) return false;
        stack<char> st;
        for(char c : s){
            if(!mp.contains(c)){ //contain得到的是key，这里取反表示若c为左括号
                st.push(c);
            }else{
                if(st.empty() || st.top() != mp[c]){ //c为右括号，此时栈为空或者栈顶不是对应的左括号都表明false
                    return false;
                }
                st.pop();   //将左括号出栈
            }
        }
        return st.empty();      //所有左括号匹配完才是true
    }
};
```

### 最小栈
设计一个支持 push ，pop ，top 操作，并能在常数时间内检索到最小元素的栈。

实现 MinStack 类:
- MinStack() 初始化堆栈对象。
- void push(int val) 将元素val推入堆栈。
- void pop() 删除堆栈顶部的元素。
- int top() 获取堆栈顶部的元素。
- int getMin() 获取堆栈中的最小元素。
 
思路：题目要求常数时间内检索到最小元素，表明栈顶元素需要保存当前栈中的最小值，因此必须使用pair<int, int> 来保存每个元素及其对应的当前最小值，相当于多用了一个辅助栈来存最小值。

注意：理解题目是关键，本题其实要求最后实现一个栈的功能，但是在实现过程中需要额外维护每个元素对应的当前最小值，从而保证 getMin() 能在常数时间内返回最小值。

实现：
```cpp
class MinStack {
    stack<pair<int, int>> st;       // 第一个存储元素值，第二个存储当前最小值
public:
    MinStack() {
        st.emplace(0, INT_MAX);     // 初始化栈，存入一个哨兵元素，保证栈不为空
    }
    
    void push(int val) {
        st.emplace(val, min(getMin(), val));    // 将元素值和当前最小值一起入栈
    }
    
    void pop() {
        st.pop();                    // 弹出栈顶元素
    }
    
    int top() {
        return st.top().first;      // 返回真实的栈顶元素值
    }
    
    int getMin() {
        return st.top().second;     // 返回栈顶元素对应的当前最小值
    }
};
```

### 字符串解码
给定一个经过编码的字符串，返回它解码后的字符串。

编码规则为: k[encoded_string]，表示其中方括号内部的 encoded_string 正好重复 k 次。注意 k 保证为正整数。

你可以认为输入字符串总是有效的；输入字符串中没有额外的空格，且输入的方括号总是符合格式要求的。

此外，你可以认为原始数据不包含数字，所有的数字只表示重复的次数 k ，例如不会出现像 3a 或 2[4] 的输入。

测试用例保证输出的长度不会超过 $10^5$。

思路：用一个栈来保护每一层的字符串和重复次数，当遇到数字时记录重复次数，当遇到 `[` 时将当前字符串和重复次数入栈，当遇到 `]` 时出栈并将当前字符串重复相应次数后拼接到栈顶字符串上。

注意：`isdigit()` 方法可以用来判断当前字符是否为数字；需要考虑多位整数的情况；`if`、`else if`和`else`的联合使用。

实现：
```cpp
class Solution {
public:
    string decodeString(string s) {
        string res = "";    //初始化ans字符串用于保存解码的答案
        long long multi = 0;
        stack<int> numStack;
        stack<string> strStack;
        for(char c : s){
            if(isdigit(c)){
                multi = multi * 10 + (c - '0'); //主要是处理多位数
            }else if(c == '['){
                numStack.push(multi);
                strStack.push(res);     // 保存已有的字符串和倍数
                multi = 0;
                res = "";
            }else if(c == ']'){
                int k = numStack.top();
                numStack.pop();
                string last_res = strStack.top();
                strStack.pop();

                string temp = "";
                for(int i = 0; i < k; i++){
                    temp += res;    // 将当前字符串重复 k 次
                }
                res = last_res + temp;
            }else{
                res += c;           // 普通字符直接加入当前字符串
            }
        }
        return res;
    }
};
```

### 每日温度
给定一个整数数组 `temperatures`，表示每天的温度，返回一个数组 `answer`，其中 `answer[i]` 是指从第 `i` 天开始，经过多少天后温度会升高。如果之后都不会升高，则在该位置用 0 来代替。

思路：每次入栈判断，如果当前温度比栈顶温度高，则说明栈顶温度找到了比它高的温度，出栈并计算下标对应的天数，接着继续判断栈顶元素，直到栈为空或者当前温度不再比栈顶温度高，然后将当前下标入栈，作为新的栈顶元素。观察栈的元素，可以发现是一个单调栈，即元素下标对应的温度是递减的。

注意：每次循环需要一直判断当前温度是否比栈顶温度高，直到栈为空或者当前温度不再比栈顶温度高为止，所以内层需要使用`while`而不是`if`。

实现：
```cpp
class Solution {
public:
    vector<int> dailyTemperatures(vector<int>& temperatures) {
        int n = temperatures.size();
        vector<int> ans(n, 0);
        stack<int> st;              //存放下标
        for(int i = 0; i < temperatures.size(); i++){
            while(!st.empty() && temperatures[i] > temperatures[st.top()]){ //将
                int pre = st.top();
                st.pop();
                ans[pre] = i - pre;
            }
            st.push(i);
        }
        return ans;
    }
};
```

### 柱状图中最大的矩形
给定一个整数数组 `heights`，表示柱状图中每个柱子的高度，返回柱状图中能够形成的最大矩形的面积。

思路：对于高度为heigth[i]的柱子，左右两边去找第一个比它矮的柱子，矩形的高度为height[i]，宽度为`right-left-1`。因此可以利用单调栈，栈用于存放下标，每次入栈时对比当前柱子的高度与栈顶柱子的高度，如果当前柱子高度小于栈顶柱子高度，则说明栈顶柱子找到了右边第一个比它矮的柱子，出栈并计算以栈顶柱子为高度的矩形面积，继续判断栈顶元素，直到栈为空或者当前柱子高度不再小于栈顶柱子高度，然后将当前下标入栈。

实现：
```cpp
class Solution {
public:
    int largestRectangleArea(vector<int>& heights) {
        int n = heights.size();
        vector<int> new_heights(n + 2, 0); //左右哨兵节点
        for(int i = 0; i < n; i++) new_heights[i + 1] = heights[i];
        stack<int> st;
        int ans = 0;
        for(int i =0; i < n + 2; i++){
            while(!st.empty() && new_heights[i] < new_heights[st.top()]){ //单调栈
                int cur_height = new_heights[st.top()];
                st.pop();
                int left = st.top();        //此时栈顶为最接近最大高度的值，即左边界
                int right = i;              //右边界即刚刚找到的小于它的值
                int width = right - left - 1;
                ans = max(ans, cur_height * width);
            }
            st.push(i);
        }
        return ans;
    }
};
```

### 数组中的第k个最大元素
给定整数数组 nums 和整数 k，请返回数组中第 k 个最大的元素。

请注意，你需要找的是数组排序后的第 k 个最大的元素，而不是第 k 个不同的元素。

你必须设计并实现时间复杂度为 O(n) 的算法解决此问题。

思路：考虑到笔试和面试需求不能使用库函数，所以先手写定义一个“快速选择”算法，通过快速排序的分区思想，每次选定一个基准，将数组划分为大于基准和小于基准两部分，然后判断第 k 个最大元素在左边还是右边，从而递归查找，这类似 二分查找，只要我们每次能把问题的规模缩小一半，就可以用 O(2n) 时间解决。写完算法后调用，一直判断索引与 n-k 的关系，从而找到第 k 个最大元素。

注意：pivot需要随机选择，使用`rand()`和`srand(time(NULL))`初始化随机种子，保证每次选择基准的随机性，从而避免最坏情况导致时间复杂度退化到 O(n^2)；`swap()`函数的使用，注意最后交换的是left和j。

实现：
```cpp
class Solution {
    int partition(vector<int>& nums, int left, int right){ //手写快速选择方法，使用双指针相向扫描
        int i = left + rand() % (right - left + 1); //随机取一个pivot
        int pivot = nums[i];
        swap(nums[i], nums[left]); //交换pivot和left

        i = left + 1;
        int j = right;
        // [ pivot | <=pivot | 尚未遍历 | >=pivot ]
        //   ^                 ^     ^         ^
        //   left              i     j         right
        while(true){
            while(i <= j && nums[i] < pivot){
                i++;
            }//此时nums[i]>=pivot
            while(i <= j && nums[j] > pivot){
                j--;
            }//此时nums[j]<=pivot
            if(i >= j){
                break;
            }
            swap(nums[i],nums[j]); //左边的大数与右边的小数交换，然后更新i和j
            i++;
            j--;
            // 循环结束后
            // [ pivot | <=pivot | >=pivot ]
            //   ^             ^   ^     ^
            //   left          j   i     right
        }
        swap(nums[left], nums[j]);
        return j;
    }
public:
    int findKthLargest(vector<int>& nums, int k) {
        srand(time(NULL));//使用当前系统时间生成真正的随机数
        int n = nums.size();
        int target_index = n - k;
        int left = 0;
        int right = n - 1;
        while(true){
            int i = partition(nums, left, right);
            if(i == target_index){
                return nums[i];
            }
            if(i > target_index){
                right = i - 1;
            }
            if(i < target_index){
                left = i + 1;
            }
        }
    }
};
```

### 前k个高频元素
给你一个整数数组 nums 和一个整数 k ，请你返回其中出现频率前 k 高的元素。你可以按 任意顺序 返回答案。

思路：先用哈希表统计每个元素出现的频率，然后将出现次数相同的元素放在一个桶内，下标代表出现次数，内容为该次数对应的元素值。最后逆向遍历桶，从出现次数最多的桶开始取元素，直到取到 k 个元素为止。

注意：桶的大小需要为 nums.size() + 1，因为元素出现次数最大可能为 nums.size()，桶下标从 0 到 nums.size()，下标 0 不会用到；入桶时可以用 `auto` 遍历哈希表，将每个元素放入对应出现次数的桶中；最后的加入数组需要考虑出现次数相同的元素，因此需要从begin到end。

实现：
```cpp
class Solution {
public:
    vector<int> topKFrequent(vector<int>& nums, int k) {
        unordered_map<int, int> cnt;    //<元素值, 出现次数>
        int max_cnt = 0;
        for(int x : nums){
            cnt[x]++;
            max_cnt = max(max_cnt, cnt[x]);
        }
        vector<vector<int>> buckets(max_cnt + 1);
        //把出现次数相同的元素放到一个桶内，下标代表频率，内容是拥有该频率的元素
        for(auto& [x, c] : cnt){ 
            buckets[c].push_back(x);
        }
        vector<int> ans;
        for(int i = max_cnt; k > ans.size(); i--){
            ans.insert(ans.end(),buckets[i].begin(),buckets[i].end());
        }
        return ans;
    }
};
```

### 数据流中的中位数
中位数是有序整数列表中的中间值。如果列表的大小是偶数，则没有中间值，中位数是两个中间值的平均值。

例如 arr = [2,3,4] 的中位数是 3 。
例如 arr = [2,3] 的中位数是 (2 + 3) / 2 = 2.5 。
实现 MedianFinder 类:
- MedianFinder() 初始化 MedianFinder 对象。
- void addNum(int num) 将数据流中的整数 num 添加到数据结构中。
- double findMedian() 返回到目前为止所有元素的中位数。与实际答案相差 $10^{-5}$ 以内的答案将被接受。

思路：将数据流中的元素分为两部分，一部分是小顶堆right存储较大的一半元素，另一部分是大顶堆left存储较小的一半元素。保证大顶堆的堆顶元素小于等于小顶堆的堆顶元素，这样中位数就是两个堆顶元素的平均值（当元素总数为偶数时）或者是堆顶元素（当元素总数为奇数时）。规定left的元素个数比right多一个或者相等，这样可以保证中位数总是出现在left的堆顶或者left和right堆顶的平均值。每次入堆时先判断元素left与right的元素个数，如果相等，则所有情况可以通过将新元素加入right然后再将right堆顶元素移到left来保持平衡；如果left比right多一个，则将新元素加入left然后再将left堆顶元素移到right来保持平衡。

注意：最后除以2的浮点数要写成 `(left.top() + right.top()) / 2.0`，否则会做整数除法导致结果不正确。

实现：
```cpp
class MedianFinder {
    priority_queue<int> left;
    priority_queue<int, vector<int>, greater<>> right;
public:    
    void addNum(int num) {
        if(left.size() == right.size()){
            right.push(num);
            left.push(right.top());
            right.pop();
        }else{
            left.push(num);
            right.push(left.top());
            left.pop();
        }
    }
    
    double findMedian() {
        if(left.size() > right.size()){
            return left.top();
        }else{
            return (left.top() + right.top()) / 2.0;
        }
    }
};
```

### 股票买卖的最佳时机
给定一个数组 prices ，其中 prices[i] 表示第 i 天的股票价格。你只能选择某一天买入这只股票，并选择在未来的某一个不同的日子卖出该股票。设计一个算法来计算你所能获取的最大利润。你不能在买入股票前卖出股票。

思路：遍历数组 prices，用一个变量记录到当前为止的最小价格 minPrice，同时用另一个变量记录最大利润 maxProfit。对于每一天的价格 prices[i]，计算如果在 minPrice 买入并在 prices[i] 卖出能获得的利润，并更新 maxProfit，同时更新 minPrice 为当前价格和 minPrice 的较小值。遍历结束后 maxProfit 即为所求最大利润。

实现：
```cpp
class Solution {
public:
    int maxProfit(vector<int>& prices) {
        int ans = 0;
        int min_price = prices[0];
        for(int p : prices) {
            ans = max(ans, p - min_price);
            min_price = min(min_price, p);
        }
        return ans;
    }
};
```

### 跳跃游戏
给定一个非负整数数组 nums ，你最初位于数组的第一个下标。数组中的每个元素代表你在该位置可以跳跃的最大长度。判断你是否能够到达最后一个下标。

思路：维护一个变量 maxReach 表示当前能够到达的最远位置，初始为 0。遍历数组 nums，对于每个位置 i，如果 i 超过了 maxReach，则说明无法到达该位置，返回 false；否则更新 maxReach 为 max(maxReach, i + nums[i])。遍历结束后，如果 maxReach 大于等于最后一个下标，则说明可以到达最后一个下标，返回 true。

注意：每次循环先判断当前位置 i 是否超过了 maxReach，如果超过则直接返回 false，否则再更新 maxReach，不能只判断最后一个位置是否小于 maxReach，因为可能在中间某个位置就已经无法到达了。

实现：
```cpp
class Solution {
public:
    bool canJump(vector<int>& nums) {
        int m = 0;
        for(int i = 0; i < nums.size(); i++){
            if(i > m) return false;
            m = max(m, i + nums[i]);
        }
        return true;
    }
};
```

### 跳跃游戏Ⅱ
给定一个非负整数数组 nums ，你最初位于数组的第一个下标。数组中的每个元素代表你在该位置可以跳跃的最大长度。你的目标是使用最少的跳跃次数到达数组的最后一个下标。假设你总是可以到达数组的最后一个下标，计算到达最后一个下标的最少跳跃次数。

思路：和“打家劫舍”不同，这里需要计算最少次数，因此每一步都需要尽量跳得远。维护两个变量，一个表示当前跳跃能够到达的最远位置cur，一个表示遍历该边界内能够到达的最远位置next。遍历数组 nums，对于每个位置 i，更新 next 为 max(next, i + nums[i])。当 i 到达 cur 时，说明必须进行一次跳跃，将 cur 更新为 next，同时跳跃次数加 1。遍历结束后跳跃次数即为最少跳跃次数。

实现：
```cpp
class Solution {
public:
    int jump(vector<int>& nums) {
        int ans = 0;
        int cur = 0;
        int next = 0;
        for(int i = 0; i < nums.size() - 1; i++){
            next = max(next, i + nums[i]); //在i处更新能到达的最远点
            if(i == cur){                   //如果现在已经走到边界了就必须跳一次
                cur = next;                 //更新cur
                ans++;                      //跳跃次数++
            }
        }
        return ans;
    }
};
```

### 划分字母区间
给定一个字符串 s ，将字符串划分为尽可能多的片段，同一字母最多出现在一个片段中，返回每个片段的长度组成的列表。

思路：如果你在当前片段里包含了一个字母 'a'，那么你的片段至少要延伸到字母 'a' 最后一次出现的位置。遍历字符串，记录far为当前所有出现的字母的最远出现位置的最大值，当遍历到i等于far时，就可以确定一个片段的结束位置，将该片段长度加入结果，并将下一个片段的起始位置更新为i+1。

注意：array直接int就行，不用`vector<int>`的类似结构。

实现：
```cpp
class Solution {
public:
    vector<int> partitionLabels(string s) {
        int last[26];                   //26个字母对应的数组
        for(int i = 0;i < s.size(); i++){
            last[s[i] - 'a'] = i;
        }                              //记录每个字母的最远距离

        vector<int> res;
        int start = 0;  //当前片段起始位置
        int far = 0;    //当前片段中所有字符能到达的最远边界
        for(int i = 0;i < s.size(); i++){
            far = max(far, last[s[i] - 'a']);
            if(i == far){
                res.push_back(i - start + 1);
                start = i + 1;
            }
        }
        return res;
    }
};
```

### 全排列
给定一个不含重复数字的数组 nums ，返回其 所有可能的全排列 。你可以 按任意顺序 返回答案。

思路：回溯法，通过递归构建全排列，用一个bool数组记录每个数字是否被使用，每次选择一个未使用的数字加入当前排列，直到排列长度等于数组长度，将其加入结果集，然后回溯撤销选择，尝试其他可能。

实现：
```cpp
class Solution {
public:
    vector<vector<int>> permute(vector<int>& nums) {
        vector<vector<int>> ans;    
        vector<int> path;    
        vector<bool> used(nums.size(), false);
        backtrack(nums, used, ans, path);
        return ans;
    }
private:
    void backtrack(const vector<int>& nums, vector<bool>& used, vector<vector<int>>& ans, vector<int>& path){
        if(path.size() == nums.size()){         //如果当前排列长度等于数组长度，则说明得到一个完整排列
            ans.push_back(path);
            return;
        }
        for(int i = 0; i < nums.size(); i++){   //used数组判断是否使用过
            if(used[i]) continue;               //如果该数字已经使用过，则跳过
            path.push_back(nums[i]);            //将该数字加入当前排列
            used[i] = true;                     //标记该数字已经使用过
            backtrack(nums, used, ans, path);   //递归调用，继续构建全排列
            path.pop_back();                    //将该数字从当前排列中移除
            used[i] = false;                    //回溯，撤销该数字的使用标记
        }
    }
};
```

### 子集
给你一个整数数组 nums ，数组中的元素 互不相同 。返回该数组所有可能的子集（幂集）。你可以按 任意顺序 返回答案。

思路：和上一题相比，子集不需要考虑元素顺序，也不需要使用 used 数组来判断元素是否被使用过，每次递归可以选择加入当前元素或者不加入当前元素，从而生成所有可能的子集。

注意：子集问题中，每个元素可以选择加入当前子集或者不加入当前元素，因此递归函数中通常会有两条分支，一条选择加入当前元素，一条选择不加入当前元素，从而生成所有可能的子集。该操作是通过`push_back()`将元素加入当前子集，然后递归调用生成子集，递归返回后通过`pop_back()`撤销选择，从而尝试不加入该元素的分支。（以\[1, 2, 3\]为例，思考\[1, 3\]是如何得到的？）

实现：
```cpp
class Solution {
public:
    vector<vector<int>> subsets(vector<int>& nums) {
        ans.clear();
        path.clear();
        backtrack(nums, 0);
        return ans;
    }
private:
    vector<vector<int>> ans;
    vector<int> path;
    void backtrack(vector<int>& nums, int startindex){
        ans.push_back(path);
        for(int i = startindex; i < nums.size(); i++){
            path.push_back(nums[i]);
            backtrack(nums, i + 1);
            path.pop_back();
        }
    }
};
```

### 电话号码的字母组合
给定一个仅包含数字 2-9 的字符串，返回所有它能表示的字母组合。答案可以按 任意顺序 返回。数字到字母的映射与电话按键相同。注意 1 不对应任何字母。

思路：回溯法，先建立数字-字母表的映射表，通过递归构建每个数字对应的字母组合（index 表示当前处理到第几个数字），每次选择当前数字对应的一个字母（i 表示当前字母在字母表中的索引）加入当前组合，然后递归处理下一个数字，直到组合长度等于输入数字长度，将其加入结果集，然后回溯撤销选择，尝试其他可能。
实现：
```cpp
class Solution {
private:
    const string map[8] = {"abc","def","ghi","jkl","mno","pqrs","tuv","wxyz"};
    vector<string> ans;
    string path;

    void backtrack(const string& digits,int index){
        if(index == digits.size()){
            ans.push_back(path);
            return;
        }

        int digit = digits[index] - '2';            //当前数字对应的字母表
        string letters = map[digit];

        for(int i = 0; i < letters.size(); i++){    //遍历当前字母表
            path.push_back(letters[i]);
            backtrack(digits, index + 1);
            path.pop_back();
        }
    }
public:
    vector<string> letterCombinations(string digits) {
        if(digits.empty()) return {};
        ans.clear();
        path.clear();
        backtrack(digits, 0);
        return ans;
    }
};
```

### 组合总和
给定一个无重复元素的数组 candidates 和一个目标数 target ，找出 candidates 中所有可以使数字和为 target 的组合。candidates 中的数字可以无限制重复被选取。答案可以按 任意顺序 返回。

思路：回溯法，通过递归构建组合，每次选择当前数字加入组合，计算sum，判断sum和target的关系，如果sum等于target，则将当前组合加入结果集，如果sum小于target，则继续递归选择下一个数字，如果sum大于target，则剪枝返回。

注意：事先`sort()`数组，这样在回溯过程中可以进行剪枝，如果当前数字大于剩余目标值，则后续数字都不可能满足条件，可以直接返回；由于数字可以重复使用，因此递归调用时传入的参数为i（表示继续取当前数字）。

实现：
```cpp
class Solution {
private:
    vector<vector<int>> ans;
    vector<int> path;
    void backtrack(vector<int>& candidates, int target, int sum, int startindex){
        if(sum == target){
            ans.push_back(path);
            return;
        }
        for(int i = startindex; i < candidates.size(); i++){
            if(sum + candidates[i] > target) break;
            path.push_back(candidates[i]);
            backtrack(candidates, target, sum + candidates[i], i);
            //传入i而不是i+1表示可以重复选同一个元素
            path.pop_back();
        }
    }
public:
    vector<vector<int>> combinationSum(vector<int>& candidates, int target) {
        ans.clear();
        path.clear();
        sort(candidates.begin(), candidates.end());
        backtrack(candidates, target, 0, 0);
        return ans;
    }
};
```

### 括号生成
数字 n 代表生成括号的对数，请你设计一个函数，用于能够生成所有可能的并且 有效的 括号组合。

思路：将题目转化为：“每次选择在当前组合中添加左括号或右括号，保证在任意时刻右括号的数量不超过左括号的数量，并且左括号的数量不超过 n，当组合长度等于 2*n 时，将其加入结果集。”

实现：
```cpp
class Solution {
private:
    // 约束条件：左括号的个数为n；起始必定为左括号->只有左括号的个数多于右括号才能放右括号
    vector<string> ans;
    void backtrack(int n,int left, int right, string& path){
        if(path.size() == 2 * n){
            ans.push_back(path);
            return;
        }
        if(left < n){
            path.push_back('(');
            backtrack(n, left + 1, right, path);
            path.pop_back();
        }
        if(right < left){
            path.push_back(')');
            backtrack(n, left, right + 1, path);
            path.pop_back();
        }
    }
public:
    vector<string> generateParenthesis(int n) {
        ans.clear();
        string path = "";
        backtrack(n, 0, 0, path);
        return ans;
    }
};
```

### 单词搜索
给定一个 m x n 二维字符网格 board 和一个字符串单词 word ，如果 word 存在于网格中，返回 true ；否则，返回 false 。

思路：二维的回溯问题，从每个格子出发，尝试向上下左右四个方向移动，匹配 word 中的字符，如果匹配成功则继续递归搜索下一个字符，同时标记当前格子已访问，回溯时恢复标记，如果搜索到 word 的最后一个字符匹配成功，则返回 true，否则继续搜索其他路径。

注意：判断条件，返回false为越界或当前格子字符不等于 word 中对应字符，返回true为已经匹配到 word 的最后一个字符；修改原数组标记当前格子已访问，回溯时恢复原数组。

实现：
```cpp
class Solution {
private:
    bool backtrack(vector<vector<char>>& board, const string& word, int k, int x, int y){
        //失败
        if(x < 0 || y < 0 || x >= board.size() || y >= board[0].size() || board[x][y] != word[k]) return false;
        //成功
        if(k == word.size() - 1) return true;
        char temp = board[x][y];
        board[x][y] = '#';
        bool res = backtrack(board, word, k + 1, x - 1, y) ||
                   backtrack(board, word, k + 1, x + 1, y) ||
                   backtrack(board, word, k + 1, x, y + 1) ||
                   backtrack(board, word, k + 1, x, y - 1);
        board[x][y] = temp;
        return res;
    }
public:
    bool exist(vector<vector<char>>& board, string word) {
        bool ans;
        int m = board.size();
        int n = board[0].size();
        for(int i = 0; i < m; i++){
            for(int j = 0; j < n; j++){
                if(backtrack(board, word, 0, i, j)) return true;
            }
        }
        return false;
    }
};
```

### 分割回文串
给你一个字符串 s，请你将 s 分割成一些 子串，使每个子串都是 回文串 。返回 s 所有可能的分割方案。

思路：回溯法，从字符串的起始位置出发，尝试每一种可能的分割方式，如果当前分割出的子串是回文串，则将其加入路径中，继续递归处理剩余的字符串，当到达字符串末尾时，将当前路径加入结果集，回溯时移除最后加入的子串，尝试其他分割方式。

注意：相比之前的回溯，本题的回溯条件是当前分割出的子串必须是回文串，只有满足这个条件才继续递归，否则直接剪枝；`substr(a, b)`的用法，b为子串的长度，而不是结束下标，所以这里要用 `i - startindex + 1` 来表示从 startindex 开始长度为 i - startindex + 1 的子串。

实现：
```cpp
class Solution {
private:
    vector<vector<string>> ans;
    vector<string> path;
    bool isPalindrome(const string& s, int start, int end){
        while(start < end){
            if(s[start++] != s[end--]) return false;
        }
        return true;
    }
    void backtrack(const string& s, int startindex){
        if(startindex == s.size()){
            ans.push_back(path);
            return;
        }
        for(int i = startindex; i < s.size(); i++){
            if(isPalindrome(s, startindex, i)){         //只有分割下来的子串是回文的才继续
                string str = s.substr(startindex, i- startindex + 1); //substc的用法
                path.push_back(str);
                backtrack(s, i + 1);
                path.pop_back();
            }
        }
    }
public:
    vector<vector<string>> partition(string s) {
        ans.clear();
        path.clear();
        backtrack(s, 0);
        return ans;
    }
};
```

### N皇后
给你一个整数 n ，返回所有不同的 n 皇后问题的解决方案。每一种解法包含一个不同的 n 皇后问题的棋子放置方案，该方案中 'Q' 和 '.' 分别代表了皇后和空位。

思路：类比“单词搜索”，从棋盘的每一行出发，尝试在每一列放置皇后，如果当前位置可以放置皇后（即不在同一列col、同一主对角线、同一副对角线已有皇后），则放置皇后并递归处理下一行，回溯时移除皇后，尝试其他列，直到放置完所有皇后，将当前棋盘状态加入结果集。用于三个bool数组usedCol、usedDiag1、usedDiag2分别记录每一列、每一主对角线、每一副对角线是否已有皇后，从而快速判断当前位置能否放置皇后。

注意：考虑副对角线时使用row - col + n来映射到usedDiag2数组的下标，保证下标非负；构建棋盘时可以用vector<string> board(n, string(n, '.'))来初始化，每次放置皇后时board[row][col] = 'Q'，回溯时board[row][col] = '.'；初始化检查数组使用array的`assign()`方法将所有元素置为false；回溯时需要将board和检查数组恢复到放置皇后之前的状态。

实现：
```cpp
class Solution {
private:
    vector<vector<string>> ans;
    vector<bool> usedCol, usedDiag1, usedDiag2;
    void backtrack(int n, int row, vector<string>& board){
        if(row == n){
            ans.push_back(board);
            return;
        }
        for(int col = 0; col < n; col++){
            if(usedCol[col] || usedDiag1[row + col] || usedDiag2[row - col + n]) continue;
            board[row][col] = 'Q';
            usedCol[col] = usedDiag1[row + col] = usedDiag2[row - col + n] = true;
            backtrack(n, row + 1, board);
            //回溯
            board[row][col] = '.';
            usedCol[col] = usedDiag1[row + col] = usedDiag2[row - col + n] = false;
        }
    }
public:
    vector<vector<string>> solveNQueens(int n) {
        ans.clear();
        vector<string> board(n, string(n, '.'));
        //初始化检查数组
        usedCol.assign(n, false);
        usedDiag1.assign(2 * n, false);
        usedDiag2.assign(2 * n, false);
        backtrack(n, 0, board);
        return ans;
    }
};
```

### 二分查找
给定一个n个元素的有序（升序）整型数组`nums`和一个目标值`target`，写一个函数搜索`nums`中的`target`，如果存在返回下标，否则返回-1。

必须编写一个具有`O(log n)`时间复杂度的算法。

思路：通过比较大小不断二分

注意：while条件；引入m，判断大小时需要±1

实现：
```cpp
#include <iostream>
#include <vector>
using namespace std;
int search(vector<int>& nums, int target){
    int n = nums.size();
    int l = 0, r = n - 1;
    while(l <= r){
        int m = l + (r - l) / 2;
        if(nums[m] == target) return m;
        else if(nums[m] < target) l = m + 1;
        else r = m - 1;
    }
    return -1;
}

int main(){
    vector<int> nums = {1, 0, 3, 5, 9, 12};
    int target = 9;
    cout << search(nums, target) << "\n";
    return 0;
}
```

### 搜索插入位置
给定一个排序数组和一个目标值，在数组中找到目标值，并返回其索引。如果目标值不存在于数组中，返回它将会被按顺序插入的位置。

请必须使用时间复杂度为 O(log n) 的算法。

思路：使用二分查找，在排序数组中查找目标值的位置。如果找到目标值，则返回其索引；如果未找到，则返回目标值应插入的位置。具体做法是维护左右指针 left 和 right，每次取中间位置 mid，如果 nums[mid] 小于目标值，则 left = mid + 1，否则 right = mid - 1。最终 left 指向的就是目标值应插入的位置。

注意：while的条件应该是 left <= right，这样才能保证当 left 超过 right 时循环结束，此时 left 就是目标值应插入的位置；需要写成`left + (right - left) / 2`防止int溢出；最后返回的就是 left。

实现：
```cpp
class Solution {
public:
    int searchInsert(vector<int>& nums, int target) {
        int left = 0;
        int right = nums.size() - 1;

        while(left <= right){
            int mid = left + (right - left) / 2;
            if(nums[mid] == target){
                return mid;
            }else if(nums[mid] > target){
                right = mid - 1;
            }else{
                left = mid + 1;
            }
        }
        return left;
    }
};
```

### 搜索二维矩阵
给你一个满足下述两条属性的 m x n 整数矩阵：
- 每行中的整数从左到右按非严格递增顺序排列。
- 每行的第一个整数大于前一行的最后一个整数。

给你一个整数 target ，如果 target 在矩阵中，返回 true ；否则，返回 false 。

思路：排除法。定义i和j分别代表第i行第j列的位置，从矩阵**右上角**开始，如果当前元素等于target，则返回true；如果当前元素小于target，说明这一行均小于target，i++；如果当前元素大于target，说明这一列均大于target，j--。

实现：
```cpp
class Solution {
public:
    bool searchMatrix(vector<vector<int>>& matrix, int target) {
        int m = matrix.size();
        int n = matrix[0].size();
        int i = 0, j = n - 1;
        while(i < m && j >= 0){
            if(matrix[i][j] == target){
                return true;
            }
            if(matrix[i][j] < target){
                i++;
            }else{
                j--;
            }
        }
        return false;
    }
};
```

### 在排序数组中查找元素的起始和结束位置
给你一个按照非递减顺序排列的整数数组 nums，和一个目标值 target。请你找出给定目标值在数组中的开始位置和结束位置。

如果数组中不存在目标值 target，返回 [-1, -1]。

你必须设计并实现时间复杂度为 O(log n) 的算法解决此问题。

思路：本质上是查找**大于等于target**的第一个元素的位置，循环结束时left恰好停在这个位置，因此开始位置可以直接取这个位置，结束位置可以通过查找**大于target**的第一个元素的位置再减一得到。

注意：返回{-1, -1}的判断条件：先判断start是否越界（即start == nums.size()，表明target比nums中所有的元素都大），再判断nums[start]是否等于target，如果不等于说明target不在数组中。

```cpp
class Solution {
    int lower_bound(vector<int>& nums, int target){
        int left = 0, right = nums.size() - 1;
        while(left <= right){
            int mid = left + (right - left) / 2;
            if(nums[mid] >= target){
                right = mid - 1;
            }else{
                left = mid + 1;
            }
        }
        return left;
    }
public:
    vector<int> searchRange(vector<int>& nums, int target) {
        int start = lower_bound(nums, target);
        if(start == nums.size() || nums[start] != target){
        //前者判断nums中的所有元素是否都小于target，后者判断target是否在target中，防止越界
            return {-1, -1};
        }
        int end = lower_bound(nums, target + 1) - 1;
        return {start, end};
    }
};
```

### 搜索旋转排序数组
整数数组 nums 按升序排列，数组中的值 互不相同 。

在传递给函数之前，nums 在预先未知的某个下标 k（0 <= k < nums.length）上进行了 向左旋转，使数组变为 [nums[k], nums[k+1], ..., nums[n-1], nums[0], nums[1], ..., nums[k-1]]（下标 从 0 开始 计数）。例如， [0,1,2,4,5,6,7] 下标 3 上向左旋转后可能变为 [4,5,6,7,0,1,2] 。

给你 旋转后 的数组 nums 和一个整数 target ，如果 nums 中存在这个目标值 target ，则返回它的下标，否则返回 -1 。

你必须设计一个时间复杂度为 O(log n) 的算法解决此问题。

思路：二分查找。由于数组是旋转排序数组，可以通过比较中间元素与左右端点的大小关系判断 target 在哪一半有序区间中，从而缩小查找范围。具体做法是每次取中间元素 mid，如果 nums[mid] == target 则返回 mid；否则判断哪一半是有序的，如果左半部分有序且 target 在左半部分范围内，则在左半部分继续查找，否则在右半部分查找；如果右半部分有序且 target 在右半部分范围内，则在右半部分继续查找，否则在左半部分查找。

注意：判断target是否在有序区间内时，需要比较 target 与有序区间的左右端点的大小关系，例如左半部分有序时判断 nums[left] <= target < nums[mid]，右半部分有序时判断 nums[mid] < target <= nums[right]。

实现：
```cpp
class Solution {
public:
    int search(vector<int>& nums, int target) {
        int left = 0;
        int right = nums.size() - 1;
        while(left <= right){
            int mid = left + (right - left) / 2;
            if(nums[mid] == target) return mid;
            if(nums[mid] > nums[right]){    //[left, mid]升序
                if(nums[left] <= target && nums[mid] > target) right = mid - 1;
                else left = mid + 1;
            }else{                          //[mid, right]升序
                if(nums[mid] < target && nums[right] >= target) left = mid + 1;
                else right = mid - 1;
            }
        }
        return -1;
    }
};
```

### 寻找排序数组中的最小值
已知一个长度为 n 的数组，预先按照升序排列，经由 1 到 n 次 旋转 后，得到输入数组。例如，原数组 nums = [0,1,2,4,5,6,7] 在变化后可能得到：
- 若旋转 4 次，则可以得到 [4,5,6,7,0,1,2]
- 若旋转 7 次，则可以得到 [0,1,2,4,5,6,7]
注意，数组 [a[0], a[1], a[2], ..., a[n-1]] 旋转一次 的结果为数组 [a[n-1], a[0], a[1], a[2], ..., a[n-2]] 。

给你一个元素值 互不相同 的数组 nums ，它原来是一个升序排列的数组，并按上述情形进行了多次旋转。请你找出并返回数组中的 最小元素 。

你必须设计一个时间复杂度为 O(log n) 的算法解决此问题。

思路：和上一题类似，通过二分查找判断中间元素与右端点的大小关系来缩小查找范围，如果中间元素大于右端点，则最小值一定在 mid 的右边，否则最小值可能是 mid 或在 mid 的左边。

注意：判断 mid 与 right 的大小关系时，如果 nums[mid] > nums[right]，说明最小值一定在 mid 的右边，否则最小值可能是 mid 或在 mid 的左边，因此 right = mid 而不是 mid - 1；while中的条件不能写成 left <= right，因为当 left == right 时，循环应该结束，此时 left 和right 都指向的是最小值的位置。

实现：
```cpp
class Solution {
public:
    int findMin(vector<int>& nums) {
        int left = 0;
        int right = nums.size() - 1;
        while(left < right){                            //这里的条件不能写等于
            int mid = left + (right - left) / 2;
            if(nums[mid] > nums[right]) left = mid + 1; //mid一定在最小值左边
            else right = mid;                           //mid可能是最小值     
        }
        return nums[left];
    }
};
```

### 寻找两个正序数组的中位数
给定两个大小分别为 m 和 n 的正序（从小到大）数组 nums1 和 nums2。请你找出并返回这两个正序数组的 中位数 。

算法的时间复杂度应该为 O(log (m+n)) 。

思路：手写几个例子观察，这道题可以理解为如何选取两个数组中的元素，组成两个数组，使第一个数组中的最大元素小于等于第二个数组中的最小元素，从而保证中位数在两个数组的分界处。因此可以枚举第一个数组中有多少个元素在左边，第二个数组中有多少个元素在左边，从而找到满足条件的分界点。

注意：边界条件的设置，例如在数组开头和结尾插入 INT_MIN 和 INT_MAX 来避免越界，同时保证在判断分界点时不会访问到数组之外的元素。

实现：
```cpp
class Solution {
public:
    double findMedianSortedArrays(vector<int>& nums1, vector<int>& nums2) {
        if(nums1.size() > nums2.size()) swap(nums1, nums2);
        int m = nums1.size();
        int n = nums2.size();
        nums1.insert(nums1.begin(), INT_MIN);
        nums1.push_back(INT_MAX);
        nums2.insert(nums2.begin(), INT_MIN);
        nums2.push_back(INT_MAX);

        int first = (m + n + 1) / 2;    
        //左边元素的个数，注意+1，表明左边比右边多一个元素（当总数为奇数时）
        int i = 0;                      //枚举nums1有i个在第一组
        int j = first - i; 
        while(true){        
        //核心是找到i,j满足：a[i] <= b[j + 1] 且 a[i + 1] >= b[j]
            if(nums1[i] <= nums2[j + 1] && nums1[i + 1] >= nums2[j]){
                int max1 = max(nums1[i], nums2[j]);
                int min2 = min(nums1[i + 1], nums2[j + 1]);
                return (m + n) % 2 ? max1 : (max1 + min2) / 2.0;
            }
            i++;
            j--;
        }
    }
};
```

## 动态规划

### 爬楼梯
假设你正在爬楼梯。需要 n 阶你才能到达楼顶。

每次你可以爬 1 或 2 个台阶。你有多少种不同的方法可以爬到楼顶呢？

思路：设dp[n]为爬n个台阶的方式，则可以将dp[n]分解为爬到第n阶的方法可以由爬到第n-1阶再迈1步上来，或者爬到第n-2阶再迈2步上来，因此有状态转移方程dp[n] = dp[n-1] + dp[n-2]。

注意：dp需要定义到n+1，因为dp[0]表示爬0阶的方法数，dp[n]表示爬到第n阶的方法数，因此数组大小至少为n+1。

实现：
```cpp
class Solution {
public:
    int climbStairs(int n) {
        if(n <= 2) return n;
        vector<int> dp(n + 1, 0);
        dp[0] = 1;
        dp[1] = 1;
        for(int i = 3; i <= n; i++){
            dp[i] = dp[i-1] + dp[i-2];
        }
        return dp[n];
    }
};
```

### 杨辉三角
给定一个非负整数 numRows，生成「杨辉三角」的前 numRows 行。

在「杨辉三角」中，每个数是它左上方和右上方的数的和。

思路：双重循环，外层循环控制行数，内层循环控制每行的每个元素，每个元素等于它左上方和右上方的元素之和。

注意：每次循环需要`resize()`。

实现：
```cpp
class Solution {
public:
    vector<vector<int>> generate(int numRows) {
        vector<vector<int>> ans(numRows);
        for(int i = 0; i < numRows; i++){
            ans[i].resize(i + 1, i);
            for(int j = 1; j < i; j++){
                ans[i][j] = ans[i-1][j-1] + ans[i-1][j];
            }
        }
        return ans;
    }
};
```

### 打家劫舍
你是一个专业的小偷，计划偷窃沿街的房屋。每间房内都藏有一定的现金，影响你偷窃的唯一制约因素就是相邻的房屋装有相互连通的防盗系统，如果两间相邻的房屋在同一晚上被小偷闯入，系统会自动报警。

给定一个代表每个房屋存放金额的非负整数数组，计算你 不触动警报装置的情况下 ，一夜之内能够偷窃到的最高金额。

思路：使用动态规划。定义 dp[i] 表示偷窃到第 i 个房屋时能够获得的最高金额，则状态转移方程为 dp[i] = max(dp[i-1], dp[i-2] + nums[i])，即偷第 i 个房屋时可以选择不偷（dp[i-1]）或者偷（dp[i-2] + nums[i]）。

实现：
```cpp
class Solution {
public:
    int rob(vector<int>& nums){
        int ans = 0;
        int n = nums.size();
        vector<int> dp(n + 1, 0);
        dp[0] = 0;
        dp[1] = nums[0];
        for(int i = 2; i <= n; i++){
            dp[i] = max(dp[i - 1], nums[i - 1] + dp[i - 2]);//注意这里是i-1
        }
        return dp[n];
    }
};
```

### 完全平方数
给定一个正整数 n，找到若干个完全平方数（比如 1, 4, 9, 16, ...）使得它们的和等于 n，并且使得所需的完全平方数的个数最少。返回这个最少的个数。

思路：定义 dp[i] 表示凑成数字 i 所需的最少完全平方数的个数。初始条件 dp[0] = 0，然后对于每个 i 从 1 到 n，尝试所有小于等于 i 的完全平方数 j\*j，状态转移方程为 dp[i] = min(dp[i], dp[i - j*j] + 1)。最终 dp[n] 就是所求的最少完全平方数的个数。

实现：
```cpp
#include <iostream>
#include <vector>
using namespace std;

class Solution {
public:
    int numSquares(int n) {
        vector<int> f(n+1);
        for(int i = 0; i <= n; i++){
            f[i] = i;
            for(int j = 1; j*j <= i; j++){
                f[i] = min(f[i], f[i - j*j] + 1);
            }
        }
        return f[n];
    }
};
 
int main()
{
    Solution sol;
    int n;
    cin >> n;
    cout << sol.numSquares(n) << endl;
    return 0;
}
```

### 零钱兑换
给你一个整数数组 coins ，表示不同面额的硬币；以及一个整数 amount ，表示总金额。计算并返回可以凑成总金额所需的 最少的硬币个数 。如果没有任何一种硬币组合能组成总金额，返回 -1 。

思路：定义 dp[i] 表示凑成金额 i 所需的最少硬币个数。初始条件 dp[0] = 0，然后对于每个 i 从 1 到 amount，尝试每个硬币面额 coin，如果 i >= coin，则状态转移方程为 dp[i] = min(dp[i], dp[i - coin] + 1)。最终 dp[amount] 就是所求的最少硬币个数，如果 dp[amount] 仍为初始值，则返回 -1。**注意初始化条件，以及内部循环判断条件，dp[i] 的初始值应设为一个大于 amount 的值以便取最小值，内部循环要判断 i >= coin 才能进行状态转移。**

实现：
```cpp
#include <iostream>
#include <vector>
using namespace std;

class Solution {
public:
    int coinChange(vector<int>& coins, int amount) {
        int n = coins.size();
        vector<int> f(amount + 1, amount + 1);
        f[0] = 0;
        for(int i = 1; i <= amount; i++){
            for(int coin : coins){
                if(i >= coin){      //以amount==1和coins[0]==1为例，必须有等号
                    f[i] = min(f[i], f[i - coin] + 1);
                }
            }
        }
        return f[amount] == amount + 1 ? -1 : f[amount];
    }
};
 
int main()
{
    Solution sol;
    int n, amount;
    cin >> n >> amount;
    vector<int> coins(n);
    for(int i = 0; i < n; i++){
        cin >> coins[i];
    }
    cout << sol.coinChange(coins, amount) << endl;
    return 0;
}
```

### 单词拆分
给你一个字符串 s 和一个字符串列表 wordDict 作为字典。如果可以利用字典中出现的一个或多个单词拼接出 s 则返回 true。

思路：定义 dp[i] 表示 s 的前 i 个字符能否拆分成字典中的单词。初始条件 dp[0] = true，然后对于每个 i 从 1 到 n，尝试每个 j 从 0 到 i，如果 dp[j] 为 true 且 s[j..i-1] 在字典中，则 dp[i] = true。最终 dp[n] 就是所求结果。

注意：用到了 unordered_set 来快速判断某个子串是否在字典中；用到了substr 来获取子串。

实现：
```cpp
#include <iostream>
#include <vector>
#include <string>
#include <unordered_set>
using namespace std;

class Solution {
public:
    bool wordBreak(string s, vector<string>& wordDict) {
        unordered_set<string> wordset(wordDict.begin(), wordDict.end());
        int n = s.length();
        // dp[i]表示s的前i个字符能否拆分
        vector<bool> dp(n + 1, false);
        dp[0] = true;
        for(int i = 1; i <= n; i++){
            // 前j个能拆分且i-j这一部分能被哈希集合找到
            for(int j = 0; j < i; j++){
                if(dp[j] && wordset.contains(s.substr(j, i-j))){
                    dp[i] = true;
                    break;
                }
            }
        }
        return dp[n];
    }
};

int main()
{
    Solution sol;
    string s;
    int n;
    cin >> s >> n;
    vector<string> wordDict(n);
    for(int i = 0; i < n; i++){
        cin >> wordDict[i];
    }
    cout << (sol.wordBreak(s, wordDict)) << endl;
    return 0;
}

```

### 最长递增子序列
给你一个整数数组 nums ，找到其中最长严格递增子序列的长度。

子序列 是由数组派生而来的序列，删除（或不删除）数组中的元素而不改变其余元素的顺序。例如，[3,6,2,7] 是数组 [0,3,1,6,2,2,7] 的子序列。

思路：dp[i] 表示以 nums[i] 这个数结尾的最长递增子序列的长度，每次循环都去看nums[i]之前的所有数nums[j]，如果 nums[j] < nums[i]，说明可以形成一个长度更长的子序列，则可以用 dp[j] 来更新 dp[i]，状态转移方程为 dp[i] = max(dp[i], dp[j] + 1)。最终答案是 dp 数组中的最大值。

实现：
```cpp
class Solution {
public:
    int lengthOfLIS(vector<int>& nums) {
        int n = nums.size();
        if(n == 0) return 0;
        vector<int> dp(n, 1);
        int ans = 1;
        for(int i = 1; i < n; i++){
            for(int j = 0; j < i; j++){
                if(nums[i] > nums[j]){  //如果nums[i]的值比j对应的大说明可以在原来的基础上增加一个子序列长度
                    dp[i] = max(dp[i], dp[j] + 1);
                }
            }
            ans = max(ans, dp[i]);
        }
        return ans;
    }
};
```

### 乘积最大子数组
给你一个整数数组 nums ，请你找出数组中乘积最大的非空连续 子数组（该子数组中至少包含一个数字），并返回该子数组所对应的乘积。测试用例的答案是一个 32-位 整数。

思路：由于是连续子数组，所以只需要考虑以每个位置和前一个位置的情况。数组元素有可能包含负数，因此需要考虑每次结尾的最大乘积和最小乘积。定义两个值分别代表现在的最大乘积和最小乘积，每次更新时根据当前元素是正数还是负数来更新这两个值。每次对比需要考虑三种情况，当前元素本身、当前元素乘以前一个位置的最大乘积、当前元素乘以前一个位置的最小乘积。以max为例，max=(当前元素，max\*当前元素，min\*当前元素)。

注意：`long long` 类型可以用来防止乘积溢出；更新max前需要保存max。

实现：
```cpp
class Solution {
public:
    int maxProduct(vector<int>& nums) {
        long long curMin = nums[0];
        long long curMax = nums[0];
        long long ans = nums[0];
        int n = nums.size();
        for(int i = 1; i < n; i++){
            long long x = nums[i];
            long long temp = curMax;
            // 因为是连续子数组，所以只需要比较上一个即可；分三种情况，之前为0；x大于0和x小于0，同时更新max和min
            curMax = max({x, temp * x, curMin * x});
            curMin = min({x, temp * x, curMin * x});

            ans = max(ans, curMax);
        }
        return ans;
    }
};
```

### 分割等和子集
给你一个 只包含正整数 的 非空 数组 nums 。请你判断是否可以将这个数组分割成两个子集，使得两个子集的元素和相等。

思路：首先判断和是否为偶数；接着可以将问题转化为“能否找到子集使得子集的和等于数组元素和的一半”。定义dp[target]表示是否存在子集的和等于target，初始dp[0] = true，对target进行递归，若dp[target - nums[i]]为true，则dp[target]也为true。

实现：
```cpp
class Solution {
public:
    bool canPartition(vector<int>& nums) {
        int sum = 0;
        for(int num : nums) sum += num;
        if(sum % 2) return false;
        int target = sum / 2;
        vector<bool> dp(target + 1, false);  // 对target进行递归
        dp[0] = true;
        for(int i = 0; i < nums.size(); i++){
            for(int j = target; j >= nums[i]; j--){
                if(dp[j - nums[i]]){
                    dp[j] = true;
                }
            }
            if(dp[target]) return true;
        }
        return dp[target];
    }
};
```

### 最长有效括号
给你一个只包含 '(' 和 ')' 的字符串，找出最长有效（格式正确且连续）括号 子串 的长度。

左右括号匹配，即每个左括号都有对应的右括号将其闭合的字符串是格式正确的，比如 "(()())"。

思路：这道题的关键是理解“有效”和“连续”，有效指的是左右括号匹配，连续指的是子串中间不能有多余的右括号打断匹配，**可以把多余的右括号看作一堵墙**，因此可以用栈记录左括号的位置，当遇到右括号时匹配栈顶的左括号，如果栈空说明遇到多余的右括号，将其作为“墙”来重新计算有效长度。

注意：这里没有使用动态规划。

实现：
```cpp
class Solution {
public:
    int longestValidParentheses(string s) {
        // 题目要求连续，因此多余的右括号是关键，将多余的右括号看作一堵墙
        stack<int> st;
        st.push(-1);
        int maxlen = 0;
        for(int i = 0; i < s.size(); i++){
            if(s[i] == '('){
                st.push(i);
            }else{
                st.pop();
                if(st.empty()){
                    st.push(i);     //栈空了，说明该右括号是那堵墙
                }
                maxlen = max(maxlen, i - st.top());
            }
        }
        return maxlen;
    }
};
```

### 不同路径
一个机器人位于一个 m x n 网格的左上角 （起始点在下图中标记为“Start” ）。机器人每次只能向下或者向右移动一步。机器人试图达到网格的右下角（在下图中标记为“Finish”）。问总共有多少条不同的路径？

思路：定义 dp[i][j] 表示从起点到达位置 (i, j) 的不同路径数，则状态转移方程为 dp[i][j] = dp[i-1][j] + dp[i][j-1]，即到达 (i, j) 的路径数等于到达 (i-1, j) 和 (i, j-1) 的路径数之和。初始条件 dp[0][0] = 1，第一行和第一列的路径数都为 1。

实现：
```cpp
class Solution {
private:
    int dp(int m, int n){
        vector<vector<int>> f(m, vector<int>(n, 1));
        for(int i = 1; i < m; i++){
            for(int j =1; j < n; j++){
                f[i][j] = f[i-1][j] + f[i][j-1];
            }
        }
        return f[m - 1][n - 1];
    }
public:
    int uniquePaths(int m, int n){
        return dp(m, n);
    }
};
```

### 最小路径和
给定一个包含非负整数的 m x n 网格 grid ，请找出一条从左上角到右下角的路径，使得路径上的数字总和为最小。

说明：每次只能向下或者向右移动一步。

思路：定义 dp[i][j] 表示从起点到达位置 (i, j) 的最小路径和，则状态转移方程为 dp[i][j] = grid[i][j] + min(dp[i-1][j], dp[i][j-1])，即到达 (i, j) 的最小路径和等于当前位置的值加上到达 (i-1, j) 和 (i, j-1) 的最小路径和中的较小值。初始条件 dp[0][0] = grid[0][0]，第一行和第一列的路径和分别为前一个位置的路径和加上当前格子的值。

注意：边界条件的处理；空间复杂度优化，直接修改原始grid。

实现：
```cpp
class Solution {
public:
    int minPathSum(vector<vector<int>>& grid) {
        int m = grid.size(), n = grid[0].size();
        for(int i = 0; i < m; i++){
            for(int j = 0; j < n; j++){
                if(i == 0 && j == 0) continue;
                if(i == 0) grid[i][j] += grid[i][j - 1];
                else if(j == 0) grid[i][j] += grid[i - 1][j];
                else grid[i][j] += min(grid[i - 1][j], grid[i][j - 1]);
            }
        }
        return grid[m - 1][n -1];
    }
};
```

### 最长回文子串
给你一个字符串 s，找到 s 中最长的 回文 子串。

思路：中心扩散法。保存左右边界的索引为left和right，遍历每个字符，尝试以该字符为中心向两边扩散，记录左右索引l和r，如果s[l] == s[r]且l、r在字符串范围内，则继续扩散，否则停止。每次扩散结束后，比较当前回文子串的长度与已记录的最长回文子串的长度，如果更长则更新left和right。

注意：奇偶判断，统一使用l =i/2 和 r=(i+1)/2；r - l - 1 表示当前回文子串的长度。

实现：
```cpp
class Solution {
public:
    string longestPalindrome(string s) {
        int n = s.size();
        int left = 0, right = 0;
        for(int i =0; i < 2 * n - 1; i++){
            int l = i / 2, r = (i + 1) / 2;
            while(l >= 0 && r < n && s[l] == s[r]){
                l--;
                r++;
            }
            if(r - l - 1 > right - left){
                left = l + 1;
                right = r - 1;
            }
        }
        return s.substr(left, right - left + 1);
    }
};
```

### 最长公共子序列
给定两个字符串 text1 和 text2，返回这两个字符串的最长 公共子序列 的长度。如果不存在 公共子序列 ，返回 0 。

一个字符串的 子序列 是指这样一个新的字符串：它是由原字符串在不改变字符的相对顺序的情况下删除某些字符（也可以不删除任何字符）后组成的新字符串。

例如，"ace" 是 "abcde" 的子序列，但 "aec" 不是 "abcde" 的子序列。
两个字符串的 公共子序列 是这两个字符串所共同拥有的子序列。

思路：动态规划，考虑 dp[i][j] 表示 text1 的前 i 个字符和 text2 的前 j 个字符的最长公共子序列长度，则状态转移方程为：如果 text1[i-1] == text2[j-1]，则 dp[i][j] = dp[i-1][j-1] + 1，否则 dp[i][j] = max(dp[i-1][j], dp[i][j-1])，初始条件全置0即可。

实现：
```cpp
class Solution {
public:
    int longestCommonSubsequence(string text1, string text2) {
        int m = text1.size(), n = text2.size();
        vector<vector<int>> dp(m + 1, vector<int> (n + 1, 0));
        //用dp[i][j]表示text1[0,...,i-1]与text2[0,...,j-1]的最大公共子序列长度
        for(int i = 1;i <= m; i++){
            for(int j = 1; j <= n; j++){
                if(text1[i - 1] == text2[j - 1]){
                    dp[i][j] = 1 + dp[i - 1][j - 1];
                }else{
                    dp[i][j] = max(dp[i - 1][j], dp[i][j - 1]);
                }
            }
        }
        return dp[m][n];
    }
};
```

### 编辑距离
给你两个单词 word1 和 word2， 请返回将 word1 转换成 word2 所使用的最少操作数  。

你可以对一个单词进行如下三种操作：
- 插入一个字符
- 删除一个字符
- 替换一个字符

思路：动态规划，定义 dp[i][j] 表示将 word1 的前 i 个字符转换成 word2 的前 j 个字符所需的最少操作数，首先边界条件为 dp[0][j] = j 和 dp[i][0] = i（分别表示把空字符串变为word2[0,...,j-1]需要j步和把word1[0,...,i-1]变为空字符串需要i步）。考虑一次操作，分为如下四种情况：
1. word1[i] == word2[j]，那么dp[i+1][j+1] = dp[i][j]，不需要操作；
2. word1[i] != word2[j]，那么可以选择三种操作中的一种：
- 替换：dp[i+1][j+1] = dp[i][j] + 1；
- 插入：dp[i+1][j+1] = dp[i+1][j] + 1；
- 删除：dp[i+1][j+1] = dp[i][j+1] + 1。

因此状态转移方程为：如果 word1[i] == word2[j]，则 dp[i+1][j+1] = dp[i][j]；否则 dp[i+1][j+1] = 1 + min({dp[i][j], dp[i+1][j], dp[i][j+1]})。

注意：dp的下标总是领先 word1 和 word2，即 dp[i+1][j+1] 对应 word1 的前 i 个字符和 word2 的前 j 个字符。

实现：
```cpp
class Solution {
public:
    int minDistance(string word1, string word2) {
        int m = word1.length(), n = word2.length();
        vector<vector<int>> f(m + 1, vector<int>(n + 1));
        //f[i][j]表示word1[0,...,i-1]变化为word2[0,...,j-1]的最少操作数
        for(int j = 0; j < n; j++){
            f[0][j + 1] = j + 1;        
        //边界条件，把空字符串变为word2[0,...,j-1]需要j步
        }
        for(int i = 0; i < m; i++){
            f[i + 1][0] = i + 1;        
        //边界条件，把word1[0,...,i-1]变为空字符串需要i步
            for(int j = 0; j < n; j++){
                if(word1[i] == word2[j]){
                    f[i + 1][j + 1] = f[i][j];
                }else f[i + 1][j + 1] = min({f[i][j], f[i+1][j], f[i][j+1]}) + 1;
            }
        }
        return f[m][n];
    }
};
```

### 岛屿数量
给你一个由 '1'（陆地）和 '0'（水）组成的的二维网格，请你计算网格中岛屿的数量。

岛屿总是被水包围，并且每座岛屿只能由水平方向和/或竖直方向上相邻的陆地连接形成。

此外，你可以假设该网格的四条边均被水包围。

思路：本质上是求孤立的岛屿的数量，如何判断“孤立”？若发现某处为“1”，他可能是单独的岛屿，也可能是某个已经发现的岛屿的一部分，因此需要通过 DFS（深度优先搜索） 或 BFS（广度优先搜索） 将与该“1”相连的所有“1”都标记为已访问，确保每个岛屿只被计数一次。采用DFS方法，遍历每个格子，若遇到“1”，计数器加一，并从该格子出发通过DFS将所有相连的“1”都标记为“0”，避免重复计数。

实现：
```cpp
class Solution {
public:
    int numIslands(vector<vector<char>>& grid) {
        int m = grid.size();
        int n = grid[0].size();
        int ans = 0;
        for(int i = 0; i < m; i++){
            for(int j = 0; j < n; j++){
                if(grid[i][j] == '1'){
                    ans++;
                    dfs(grid, i, j);
                }
            }
        }
        return ans;
    }
private:
    void dfs(vector<vector<char>>& grid, int row, int col){
        if(row < 0 || row >= grid.size() || col < 0 || col >= grid[0].size() || grid[row][col] == '0')  return;
        grid[row][col] = '0';
        dfs(grid, row + 1, col);
        dfs(grid, row - 1, col);
        dfs(grid, row, col - 1);
        dfs(grid, row, col + 1);
    }
};
```

### 腐烂的橘子
在给定的 m x n 网格 grid 中，每个单元格可以有以下三个值之一：
- 值 0 代表空单元格；
- 值 1 代表新鲜橘子；
- 值 2 代表腐烂的橘子。

每分钟，腐烂的橘子 周围 4 个方向上相邻 的新鲜橘子都会腐烂。

返回 直到单元格中没有新鲜橘子为止所必须经过的最小分钟数。如果不可能，返回 -1 。

思路：“岛屿数量”只关心岛屿的连通性，而“腐烂的橘子”关心的是最短时间，因此不能用DFS而是得用BFS，这题的关键是如何快速判断哪些新鲜橘子会在下一分钟被腐烂橘子感染，继续向外扩散，直到没有新鲜橘子为止。可以用一个队列来存放当前所有腐烂橘子的坐标，并记录minutes和fresh，使用while循环表示一直扩散的过程，终止条件为队列为空或fresh为0，每次循环minutes++，并处理当前队列中的所有腐烂橘子，先出队然后扩散到四个方向上相邻的新鲜橘子，将它们标记为腐烂并加入队列，同时fresh--，循环结束后判断fresh是否为0，为0则返回minutes，否则返回-1。

注意：while终止条件；if判断条件；if判断后记得入队。

实现：
```cpp
class Solution {
public:
    int orangesRotting(vector<vector<int>>& grid) {
        int m = grid.size(), n = grid[0].size();
        queue<pair<int, int>> q;
        int minutes = 0;
        int fresh = 0;
        for(int i = 0; i < m; i++){ //遍历得到初始的腐烂橘子和新鲜橘子数量
            for(int j = 0; j < n; j++){
                if(grid[i][j] == 2) q.push({i, j});
                else if(grid[i][j] == 1) fresh++;
            }
        }
        if(fresh == 0) return 0;
        int directions[4][2] = {{1, 0}, {-1, 0}, {0, 1}, {0, -1}};
        while(!q.empty() && fresh > 0){        //终止条件为队列为空或fresh为0
            minutes++;
            int size = q.size();
            for(int i = 0; i < size; i++){      //处理当前这一分钟的腐烂橘子，弹出队列
                auto [r, c] = q.front();
                q.pop();
                for(auto& d : directions){
                    int rc = r + d[0], cc = c + d[1];
                    if(rc < m && rc >= 0 && cc < n && cc >= 0 && grid[rc][cc] == 1){
                        //经过这个判断只会剩下最外围的腐烂橘子，然后进行下一轮
                        grid[rc][cc] = 2;
                        fresh--;
                        q.push({rc, cc});
                    }
                }
            }
        }
        return fresh == 0 ? minutes : -1;
    }
};
```

### 课程表
你这个学期必须选修 numCourses 门课程，记为 0 到 numCourses - 1 。

在选修某些课程之前需要一些先修课程。 先修课程按数组 prerequisites 给出，其中 prerequisites[i] = [ai, bi] ，表示如果要学习课程 ai 则 必须 先学习课程  bi 。

- 例如，先修课程对 [0, 1] 表示：想要学习课程 0 ，你需要先完成课程 1 。

请你判断是否可能完成所有课程的学习？如果可以，返回 true ；否则，返回 false 。

思路：本题本质上是判断有向图中是否存在环。可以将课程和先修课程看作有向图中的节点和有向边，若存在环，则说明存在课程依赖循环，无法完成所有课程；若不存在环，则可以完成所有课程。用adj来表示每个节点指向的节点，用indegree来统计每个节点的入度，先将入度为 0 的节点入队，然后依次出队并将其指向的节点入度减 1，若最终所有节点都被访问过，则不存在环，否则存在环。

注意：题目中的prerequisites[i] = [ai, bi] 表示bi -> ai，表现在adj中是 adj[bi].push_back(ai)；容易混淆adj数组的顺序，可以实际写一下小例子验证一下方向是否正确。

实现：
```cpp
class Solution {
public:
    bool canFinish(int numCourses, vector<vector<int>>& prerequisites) {
        vector<vector<int>> adj(numCourses);    //创建有向图,[a, b]表示b -> a
        vector<int> inDegree(numCourses, 0);    //创建度
        for(auto& pre : prerequisites){
            adj[pre[1]].push_back(pre[0]);      //pre[1]的后续课程为pre[0]
            inDegree[pre[0]]++;                 //更新pre[0]的度（指向其的箭头）
        }
        queue<int> q;
        for(int i = 0; i < numCourses; i++){
            if(inDegree[i] == 0) q.push(i);      //入队所有度为0的课程
        }
        int count = 0;
        while(!q.empty()){
            int cur = q.front();
            q.pop();
            count++;
            for(int next : adj[cur]){
                inDegree[next]--;
                if(inDegree[next] == 0) q.push(next);
            }
        }
        return count == numCourses;
    }
};
```

### 实现Trie（前缀树）
Trie（发音类似 "try"）或者说 前缀树 是一种树形数据结构，用于高效地存储和检索字符串数据集中的键。这一数据结构有相当多的应用情景，例如自动补全和拼写检查。

请你实现 Trie 类：
- Trie() 初始化前缀树对象。
- void insert(String word) 向前缀树中插入字符串 word 。
- boolean search(String word) 如果字符串 word 在前缀树中，返回 true（即，在检索之前已经插入）；否则，返回 false 。
- boolean startsWith(String prefix) 如果之前已经插入的字符串 word 的前缀之一为 prefix ，返回 true ；否则，返回 false 。

思路：类似二叉树，这里用一个children数组表示26个字母，每个节点有一个长度为26的children数组表示它的子节点，同时用一个bool变量表示该节点是否是某个单词的结尾。“插入”从根节点开始，依次处理每个字符，如果对应的children为空则新建节点，最后将最后一个字符对应的节点标记为单词结尾。“查找”从根节点开始依次查找每个字符，如果中途遇到为空的children则返回false，否则查找到最后一个字符时根据是否是单词结尾返回结果。“前缀”从根节点开始依次查找每个字符，如果中途遇到为空的children则返回false，否则查找到最后一个字符时直接返回true，不需要判断是否是单词结尾。

实现：
```cpp
class Trie {
private:
    struct TrieNode{
        TrieNode* children[26];
        bool isEnd;
        TrieNode(){
            for(int i = 0; i < 26; i++) children[i] = nullptr;
            isEnd = false;
        }
    };

    TrieNode* root;
public:
    Trie() {
        root = new TrieNode();    
    }
    
    void insert(string word) {
        TrieNode* node = root;
        for(char c : word){
            int i = c - 'a';
            if(node->children[i] == nullptr){
                node->children[i] = new TrieNode();
            }
            node = node->children[i];
        }
        node->isEnd = true;
    }
    
    bool search(string word) {
        TrieNode* node = root;
        for (char c : word){
            int i = c - 'a';
            if (node->children[i] == nullptr) return false;
            node = node->children[i];
        }
        return node->isEnd;
    }
    
    bool startsWith(string prefix) {
        TrieNode* node = root;
        for(char c : prefix){
            int i = c - 'a';
            if(node->children[i] == nullptr) return false;
            node = node->children[i];
        }
        return true;
    }
};
```

### 只出现一次的数字
给你一个 非空 整数数组 nums ，除了某个元素只出现一次以外，其余每个元素均出现两次。找出那个只出现了一次的元素。

你必须设计并实现线性时间复杂度的算法来解决此问题，且该算法只使用常量额外空间。

思路：异或`^`的性质是“相同为0， 不同为1”。
1. 一个数和自己异或结果为0，即 a ^ a = 0
2. 一个数和0异或结果为自己，即 a ^ 0 = a
3. 异或运算满足交换律和结合律

因此，对于数组中出现两次的数字，异或后会抵消为0，最终结果就是只出现一次的数字。

实现：
```cpp
class Solution {
public:
    int singleNumber(vector<int>& nums) {
        int ans = 0;
        for(int i = 0; i < nums.size(); i++){
            int x = nums[i];
            ans = ans ^ x;
        }
        return ans;
    }
};
```

### 多数元素
给定一个大小为 n 的数组 nums ，返回其中的多数元素。多数元素是指在数组中出现次数 大于 ⌊ n/2 ⌋ 的元素。

你可以假设数组是非空的，并且给定的数组总是存在多数元素。

思路1：排序后众数一定出现在中间位置。

实现1：
```cpp
class Solution{
public:
    int majorityElement(vector<int>& nums) {
        sort(nums.begin(), nums.end());
        return nums[nums.size() / 2];
    }
};
```

思路2：哈希表存储并在遍历时比较大小。

实现2：
```cpp
class Solution{
public:
    int majorityElement(vector<int>& nums) {
        unordered_map<int, int> mp;
        int majority = 0, maxcount = 0;
        for(int num : nums){
            mp[num]++;
            if(mp[num] > maxcount){
                maxcount = mp[num];
                majority = num;
            }
        } 
        return majority;
    }
};
```

### 颜色分类
给定一个包含红色、白色和蓝色、共 n 个元素的数组 nums ，**原地**对它们进行排序，使得相同颜色的元素相邻，并按照红色、白色、蓝色顺序排列。

我们使用整数 0、 1 和 2 分别表示红色、白色和蓝色。

必须在不使用库内置的 `sort` 函数的情况下解决这个问题。

思路1：遍历之后统计每种颜色的个数然后逐一填充数组。

实现1：
```cpp
class Solution {
public:
    void sortColors(vector<int>& nums) {
        int map[3] = {0, 0, 0};
        for(int num : nums){
            map[num]++;
        }
        int index = 0;
        for(int i = 0; i < 3; i++){
            for(int j = 0; j < map[i]; j++){
                nums[index++] = i;
            }
        }
    }
};
```

思路2：设red、white、blue分别代表红色、白色和蓝色子数组的尾部，排序后的数组具有天然的关系“red<=white<=blue”。每次遍历到一个数后进行一次“扩张”，遇到蓝色(2)，blue++并赋值2；遇到白色(white)，blue++并赋值2，然后 white++ 并赋值1；遇到红色(0)，blue++并赋值2，white++并赋值1，red++并赋值0。为什么不会影响到后面的元素？每次遍历都保证了red<=white<=blue<=i。

实现2：
```cpp
class Solution{
public:
    void sortColors(vector<int>& nums) {
        int red = 0, white = 0, blue = 0;
        for(int i = 0; i < nums.size(); i++){
            if(nums[i] == 0){
                nums[blue++] = 2;
                nums[white++] = 1;
                nums[red++] = 0;
            }else if(nums[i] == 1){
                nums[blue++] = 2;
                nums[white++] = 1;
            }else{
                nums[blue++] = 2;
            }
        }
    }
};
```

### 下一个排列
整数数组的一个 排列  就是将其所有成员以序列或线性顺序排列。

- 例如，arr = [1,2,3] ，以下这些都可以视作 arr 的排列：[1,2,3]、[1,3,2]、[3,1,2]、[2,3,1] 。

整数数组的 下一个排列 是指其整数的下一个字典序更大的排列。更正式地，如果数组的所有排列根据其字典顺序从小到大排列在一个容器中，那么数组的 下一个排列 就是在这个有序容器中排在它后面的那个排列。如果不存在下一个更大的排列，那么这个数组必须重排为字典序最小的排列（即，其元素按升序排列）。

- 例如，arr = [1,2,3] 的下一个排列是 [1,3,2] 。
- 类似地，arr = [2,3,1] 的下一个排列是 [3,1,2] 。
- 而 arr = [3,2,1] 的下一个排列是 [1,2,3] ，因为 [3,2,1] 不存在一个字典序更大的排列。

给你一个整数数组 nums ，找出 nums 的下一个排列。

必须**原地**修改，只允许使用额外常数空间。

思路：核心是从右往左找，找到第一个破坏递增规律的数（可以自己写一个例子验证一下）。具体可以分为三步走：
- 从右往左找第一个i，使得a[i] < a[i+1]；
- 确定i后确定最小的比a[i]大的数a[j]并交换a[i]和a[j];
- 将现在a[j]之后的元素按升序排列

注意：while判断时的等号选择

实现：
```cpp
class Solution(){
public:
    void nextPermutation(vector<int>& nums) {
        int n = nums.size();
        int i = n - 2;
        while(i >= 0 && nums[i] >= nums[i + 1]){
            i--;
        }
        if(i >= 0){
            int j = n - 1;
            while(j >=0 && nums[j] <= nums[i]){             //找到第一个比a[i]大的a[j]
                j--;
            }
            std::swap(nums[i], nums[j]);
        }
        std::reverse(nums.begin() + i + 1, nums.end());     //包含最大字典序的情形
    }
};
```

### 寻找重复数
给定一个包含 n + 1 个整数的数组 nums ，其数字都在 [1, n] 范围内（包括 1 和 n），可知至少存在一个重复的整数。

假设 nums 只有 一个重复的整数 ，返回 这个重复的数 。

你设计的解决方案必须 不修改 数组 nums 且只用常量级 O(1) 的额外空间。

思路：参考“环形链表Ⅱ”，将下标i作为当前节点，nums[i]作为该节点指向的下一个节点的地址。对于环形链表Ⅱ，做法是快慢指针，相遇后将快指针置0，然后两者速度相同至相遇。

实现：
```cpp
class Solution {
public:
    int findDuplicate(vector<int>& nums) {
        int slow = 0, fast = 0;
        while(true){
            slow = nums[slow];
            fast = nums[nums[fast]];
            if(slow == fast) break;
        }
        fast = 0;
        while(slow != fast){
            slow = nums[slow];
            fast = nums[fast];
        }
        return slow;
    }
};
